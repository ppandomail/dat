## Indicadores de calidad (DQI)

**DQI (Data Quality Indicator)**: es una métrica cuantificable que refleja el estado de una dimensión de calidad
* **Ejemplos concretos**
    * % de valores nulos
    * % de duplicados
    * % de registros válidos
    * % de consistencia entre sistemas

$$Completitud = \dfrac{valores \; no \; nulos}{total \; de \; valores}$$
$$Duplicados = \dfrac{registros \; duplicados}{total \; de \; registros}$$

**Componentes de un DQI**:
* **Dimensión de calidad**: Ej: completitud, exactitud, consistencia
* **Definición formal**: Qué mide exactamente
* **Fórmula de cálculo**: Cómo se cuantifica
* **Umbral (threshold)**: Qué valor es aceptable
* **Contexto de uso**: Para qué proceso/decisión sirve

**Ejemplo**: **DQI: Completitud de DNI en pacientes**
* **Dimensión**: completitud
* **Definición**: proporción de registros con DNI informado
* **Fórmula**: $\dfrac{registros \; con \; DNI}{total \; de \; registros}$
* **Umbral**: >= 98%
* **Contexto**: identificación única de pacientes

**Dimensiones de calidad más utilizadas**

| Dimensión        | Qué mide                       | Ejemplos de DQI      |
| --               | --                             | --                   |
| **completitud**  | datos faltantes                | % nulos              |
| **exactitud**    | corrección del dato            | % valores correctos  |
| **consistencia** | coherencia entre datos         | % reglas cumplidas   |
| **validez**      | cumplimiento de formato/reglas | % valores válidos    |
| **unicidad**     | no duplicación                 | % registros únicos   |
| **actualidad**   | vigencia temporal              | % datos actualizados |

**Alineación con negocio**
* **Técnico**: % nulos en email = 20%
* **Negocio**: 20% de clientes no pueden ser contactados -> impacto en campañas

**Ciclo de vida de un DQI**
1. Definición
2. Implementación
3. Medición periódica
4. Monitoreo
5. Acción (corrección)

**Sin acción -> no hay gestión de calidad**

## Diseño de métricas alineadas al negocio

**No todas las métricas importan: importan las que impactan en decisiones**

| Ejemplo      | Problema                     | Métrica relevante                                                  |
| --           | --                           | --                                                                 |
| **Salud**    | errores en datos de donantes | % de datos completos en variables críticas (grupo sanguíneo, edad) |
| **RRHH**     | mala calidad en CVs          | % de perfiles con información completa para selección              |
| **Finanzas** | inconsistencias en reportes  | % de registros consistentes entre fuentes                          |

## Visualización de resultados con dashboards

**Un dashboard NO es solo gráfico -> es toma de decisión**
* **Buenas prácticas**:
    * Simplicidad
    * KPIs claros
    * Uso de colores (alertas)
    * Evolución temporal

**Ejemplos de dashboard de calidad**
* % completitud (gauge o barra)
* % duplicados
* tendencia de calidad en el tiempo
* alertas (rojo/verde)

**¿Qué debería tener un dashboard de calidad?**
* KPIs claros -> completitud, duplicados
* Evolución temporal -> tendencias en el tiempo
* Alertas -> soporte a decisión
* Interpretación (no solo gráfico)
* Umbrales -> toma de decisión

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Dataset simulado
data = {
    'fecha': pd.date_range(start='2024-01-01', periods=10, freq='ME'),
    'total_registros': [100, 120, 110, 130, 125, 140, 150, 160, 155, 170],
    'nulos': [10, 15, 8, 20, 18, 25, 30, 28, 26, 35],
    'duplicados': [5, 6, 4, 7, 6, 8, 10, 9, 8, 11]
}

df = pd.DataFrame(data)
```

```python
# 2. KPIs de calidad
df['completitud'] = 1 - (df['nulos'] / df['total_registros'])
df['pct_duplicados'] = df['duplicados'] / df['total_registros']

print('Dataset con métricas de calidad')
display(df)
```

```python
# 3. KPI actuales (último período)
ultimo = df.iloc[-1]

print('KPIs actuales:')
print(f'Completitud: {ultimo['completitud']:.2%}')
print(f'% Duplicados: {ultimo['pct_duplicados']:.2%}')
```

```python
# 4. Dashboard (gráficos)
plt.figure()
plt.plot(df['fecha'], df['completitud'], marker='o')
plt.title('Evolución de Completitud')
plt.xlabel('Fecha')
plt.ylabel('Completitud')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
plt.figure()
plt.plot(df['fecha'], df['pct_duplicados'], marker='o')
plt.title('Evolución de % Duplicados')
plt.xlabel('Fecha')
plt.ylabel('% Duplicados')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
# 5. Alertas simples
umbral_completitud = 0.9
umbral_duplicados = 0.1

print('🚨 Alertas:')

if ultimo['completitud'] < umbral_completitud:
    print('⚠️ Baja completitud detectada')
else:
    print('✅ Completitud OK')

if ultimo['pct_duplicados'] > umbral_duplicados:
    print('⚠️ Alto nivel de duplicados')
else:
    print('✅ Duplicados OK')
```

### ¿Por qué visualizar la calidad de datos?

* Medir calidad sin visualizarla limita la toma de decisiones.
* Los dashboards permiten transformar métricas técnicas en información clara para responsables de negocio, líderes de áreas y equipos operativos.

### Ejemplos de dashboards

### ¿Por qué visualizar la calidad de datos?

Un dashboard de Data Quality debe responder rápidamente preguntas como:
* ¿Qué tan confiables son nuestros datos hoy?
* ¿Qué procesos están generando errores?
* ¿Dónde hay mayores riesgos?
* ¿La calidad mejora o empeora con el tiempo?
* ¿Qué áreas requieren intervención prioritaria?

### Beneficios de utilizar dashboards

* Monitoreo continuo de calidad.
* Identificación temprana de incidentes.
* Priorización de acciones correctivas.
* Comunicación entre áreas técnicas y negocio.
* Seguimiento de KPIs y SLA de datos.
* Auditoría y trazabilidad.

### Métricas frecuentes para visualizar

| Métrica          |                                                               | Ejemplo                                       |
| --               | --                                                            | --                                            |
| **Exactitud**    | Porcentaje de registros correctos respecto a fuente confiable | Clientes con DNI válido: 97%                  |
| **Completitud**  | Campos obligatorios completos                                 | Emails cargados: 88%                          |

### Tipos de visualizaciones recomendadas

Un **KPI** (del inglés Key Performance Indicator), o **indicador clave de desempeño**, es una métrica específica que permite medir el éxito y el progreso de una empresa, proyecto o estrategia respecto a sus objetivos establecidos 

**Gráficos de tendencia**
* Permiten ver evolución semanal o mensual
* Los gráficos de tendencia son la herramienta visual estándar para analizar KPIs a lo largo del tiempo
* Su función principal es transformar datos aislados en una narrativa que revela si vas por buen camino o si necesitas ajustar la estrategia

**Barras comparativas**
* Comparar áreas, sucursales, sistemas o proveedores.
* ¿Para qué sirven?
    * Identificar Patrones: Detectar estacionalidad (ej. subas cada diciembre).
    * Monitorear KPIs: Ver si una métrica se acerca o se aleja de una meta mensual.
    * Hacer Proyecciones: Ayudan a predecir comportamientos futuros basados en el historial

**Heatmaps**
* Detectar zonas críticas rápidamente.
* Los Heatmaps o mapas de calor son representaciones visuales que usan una escala de colores para mostrar la intensidad de los datos en una matriz o espacio.
* Son la herramienta definitiva para detectar zonas críticas (puntos calientes) de un solo vistazo, sin
necesidad de analizar tablas numéricas complejas.
* **¿Cómo detectan las zonas críticas?**
* Funcionan bajo una lógica térmica:
    * **Zonas Rojas/Cálidas**: Indican alta densidad, frecuencia o riesgo. Son los puntos donde se concentra la acción (clics, tráfico o peligro).
    * **Zonas Azules/Frías**: Indican áreas de bajo interés, poca actividad o baja probabilidad

**Tablas operativas**
* Detalle de errores para acción inmediata.
* El objetivo de esta tabla es la resolución rápida. Debe mostrar solo lo que está "roto" o fuera de proceso para evitar el ruido visual.

**Semáforos (Rojo/Amarillo/Verde)**
* Estado general según umbrales. 
* Ejemplo: Aquí se resumen miles de datos en un solo indicador visual que cambia según tus reglas de negocio.
* Interpretación:
    * **Verde (95%)**: Todo bajo control. No requiere intervención
    * **Amarillo (78%)**: Cerca del límite permitido. Requiere observación
    * **Rojo (45%)**: Crisis. Requiere atención prioritaria del gerente

### Buenas prácticas de diseño

* Mostrar pocas métricas clave
* Priorizar claridad sobre complejidad
* Usar colores consistentes
* Incluir filtros por fecha/área
* Mostrar tendencia histórica
* Permitir drill-down al detalle
* Definir responsables por indicador
