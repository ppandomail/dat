# Fundamentos de la calidad de datos
---

## Calidad de Software

* Es el grado en que un sistema, componente o proceso de software **cumple con los requisitos** especificados, **necesidades o expectativas** del usuario y/o del cliente (ISO/IEC 25010:2011)
* La concordancia con los requisitos funcionales y de rendimiento explícitamente establecidos, con los estándares de desarrollo explícitamente documentados y con las características implícitas que se espera de todo software profesionalmente desarrollado (IEEE Std 610.12-1990)
* La **satisfacción del usuario** con respecto a cómo el software cumple sus necesidades, incluidas sus expectativas de
funcionalidad, rendimiento, fiabilidad, facilidad de uso, facilidad de aprendizaje, seguridad y capacidad de
modificación" (Boehm, 1981)

## Algunos aspectos clave que influyen en la calidad del software

## ¿Por qué es fundamental mantener la calidad de software?

* La calidad del software es fundamental, ya que un **software de baja calidad puede llevar a errores costosos, problemas de seguridad y descontento de los usuarios**
* Por otro lado, un software de alta calidad puede generar una mayor satisfacción del cliente, reducir costos operativos y mejorar la productividad general
* Para garantizar y mejorar la calidad del software, se aplican diversas prácticas y metodologías de desarrollo, como pruebas exhaustivas, revisión de código, automatización de pruebas, seguimiento de métricas de calidad y uso de estándares de desarrollo reconocidos

## Datos

* Pueden ser de diferentes tipos y formatos, como números, texto, imágenes, archivos, registros de bases de datos, entre otros

* Los datos son fundamentales en el desarrollo y funcionamiento de cualquier software, ya que permiten que el programa realice cálculos, tome decisiones, almacene información, interactúe con los usuarios y genere resultados

## Calidad de datos

* Se da cuando los datos se ajustan al propósito para el que fueron destinados
* Los datos también se consideran de alta calidad cuando representan con precisión los conceptos del mundo real
* Los datos son la piedra angular de una jerarquía que se construye sobre ellos
* Sobre la base de los datos, viene la información, que son datos colocados en contexto
* De la **información** procesable surge el **conocimiento**, que se convierte en sabiduría cuando se aplica

## Problemas con calidad de datos

* Los datos de mala calidad darán como resultado una mala calidad de la información, y esto sube en la jerarquía, lo que da como resultado malas decisiones comerciales

## Ciclo de vida del dato

* Describe las diferentes etapas por las que atraviesa un dato:
    1. **Creación y Captura**: desde diversas fuentes, como sensores, sistemas de registro, formularios en línea o
interacciones de usuarios
    2. **Almacenamiento y Respaldo**: en bases de datos, sistemas de archivos o almacenamiento en la nube para su posterior acceso y uso
    3. **Procesamiento y Uso**: para extraer información significativa, realizar análisis, generar informes y apoyar la toma de decisiones
    4. **Integración y Calidad de Datos**: es necesario integrarlos para garantizar la consistencia y calidad de los mismos
    5. **Compartir y Distribuir**: se comparten con otras partes interesadas, equipos o sistemas para su utilización en diferentes contextos
    6. **Mantenimiento y Actualización**: se llevan a cabo acciones para mantener los
datos actualizados y precisos a lo largo del tiempo
    7. **Privacidad y Seguridad**
    8. **Archivo y Eliminación**: Los datos que ya no son necesarios pueden archivarse para referencia futura o eliminarse de manera segura según las políticas de retención de datos

## Beneficios del ciclo de vida del dato

* Mejora de procesos
* Control de costos
* Usabilidad de los datos
* Accesibilidad de los datos

## Gestión del ciclo de vida de los datos (DLM)

* Es un enfoque para gestionar los datos en su ciclo de vida, desde la entrada hasta la destrucción de los datos
* Los datos se separan en fases en función de diferentes criterios y se mueven a través de estas etapas a medida que completan diferentes tareas o cumplen ciertos requisitos
* **Un buen proceso de DLM proporciona estructura y organización a los datos de una empresa, lo que a su vez permite
objetivos principales dentro del proceso, como la seguridad y la disponibilidad de los datos**

## Ciclo de vida de los datos (otra fuente)

1. **Generación y captura**:
    * Captura Manual (formularios u hojas de cálculo)
    * Sensores
    * Web Scraping: extraer datos de páginas web
    * Integración de Sistemas: (APIs, ETL)
    * Captura Automatizada de Registos. Registros de red o tráfico de dato
1. **Almacenamiento**: en sistemas o bases de datos
    * Datos no estructurados (Imágenes, textos, PDF)
    * Datos estructurados (BBDD Relacionales, CSV)
    * Datos semi estructurados (XML, JSON)
1. **Tratamiento**:
    * **Limpieza de Datos**: se identifican y corrigen los errores, valores atípicos y datos faltantes. Eliminar duplicados
    * **Transformación de Datos**: normalización de datos, conversión de unidades de medida, la codificación de variables categóricas y la agregación de datos en diferentes niveles de granularidad
    * **Control de Calidad de Datos**: se establecen criterios y métricas para evaluar: integridad, consistencia y la exactitud
    * **Segmentación de Datos**: se crean subconjuntos para un análisis más detallado
    * **Gestión de Cambios**: establecer procesos para gestionar cambios en los datos y garantizar que la información se mantenga actualizada
1. **Análisis**:
    * **Exploración de Datos**: generación de gráficos, tablas y resúmenes estadísticos para poder comprender mejor la distribución de los datos y evaluar la correlación entre variables
    * **Análisis Estadístico**: para comprender la variabilidad de los datos y realizar inferencias sobre poblaciones o fenómenos. Se puede realizar pruebas de hipótesis, análisis de varianza o regresiones para obtener información significativa
    * **Aprendizaje Automático (ML)**: se utilizan algoritmos de aprendizaje automático para crear modelos predictivos y descriptivos. Esto incluye clasificación, regresión, agrupación y tareas de procesamiento del lenguaje natural (NLP). Las herramientas populares incluyen scikit-learn, TensorFlow, y PyTorch
1. **Visualización**: comunicar de manera efectiva la información a los interesados y tomar decisiones informadas
    * **Selección de Tipos de Gráficos**: gráficos de barras, circulares, de dispersión hasta mapas de calor o líneas de tiempo. Esta elección depende de la naturaleza de los datos
    * **Diseño de Gráficos Efectivos**: deben ser claros, legibles y atractivos. Se deben elegir colores, etiquetas y títulos de manera cuidadosa para una compresión fácil y rápida
    * **Tableros de Control (Dashboards)**: conjunto de visualizaciones que se utilizan para mostrar un panorama completo de los datos y las métricas. Son útiles para monitoreo a tiempo real o toma de decisiones empresariales
    * **Herramientas**: Tableau, Power BI, Matplotlib, Seaborn, D3.js

## Características de calidad de datos

* Los datos poseen diversas características que describen sus atributos y cualidades
* Son métricas utilizadas para evaluar si los datos cumplen con los requisitos de negocio: 

| Dimensiones                   |    |
| --                            | -- |
| **Precisión (Exactitud)**     | qué tan cercanos están los valores a la realidad o al valor verdadero. Datos precisos son fundamentales para obtener resultados confiables y tomar decisiones informadas |
| **Complejidad**               | los datos pueden variar en su complejidad, desde valores simples como números enteros o caracteres hasta estructuras más complicadas como bases de datos con múltiples tablas y relaciones |
| **Tipo**                      |  los datos pueden clasificarse en diferentes tipos, como datos numéricos (enteros, decimales), datos alfanuméricos (texto), datos binarios (0 y 1) o datos multimedia (imágenes, audio, video) |
| **Volumen**                   | cantidad de información que se encuentra disponible |
| **Velocidad**                 | se refiere a la rapidez con la que los datos son generados, procesados y transmitidos |
| **Veracidad**                 | referencia a la confiabilidad y autenticidad de los datos. Es importante que los datos sean veraces y provengan de fuentes confiables para evitar la propagación de información incorrecta o falsa |
| **Consistencia**              | se refiere a que los datos deben mantener un formato uniforme y coherente en toda su estructura para asegurar su correcta interpretación y uso |
| **Actualidad**                | indica qué tan recientes son y si siguen siendo válidos para el propósito actual. En ciertas situaciones, los datos más antiguos pueden haber perdido relevancia o ser menos útiles |
| **Accesibilidad**             | facilidad con la que se puede acceder a los datos. La disponibilidad de datos juega un papel crucial en la toma de decisiones informadas |

## Ciencia de datos

* Es un campo interdisciplinar que haciendo uso del método científico extrae conocimiento de un conjunto de datos disponible
* Se enfoca en el estudio y análisis para obtener información significativa, conocimiento y dominio del tema para transformar los datos crudos en información valiosa
* Es el proceso de transformar datos crudos en decisiones accionables mediante el uso de análisis estadístico y computacional
* Descubrir el "porque" detrás de los datos

## Científico de datos

* Es el arquitecto de la inteligencia de datos, quien se sumerge profundamente en los conjuntos de datos, desarrolla modelos avanzados y utiliza algoritmos de aprendizaje automático para predecir tendencias y patrones futuros

## MDM (Master Data Management)

**La gestión de datos maestros (MDM) surgió de la necesidad de las empresas de mejorar la coherencia y la calidad de sus activos de datos clave, como datos de productos, datos de activos, datos de clientes, datos de ubicación, etc**

* Es un conjunto de procesos y tecnologías utilizadas para garantizar la calidad, coherencia, integridad y disponibilidad de los datos maestros en una organización

* Los **datos maestros** son aquellos datos fundamentales y comunes que describen las entidades principales con las que una empresa opera, como clientes, proveedores, productos, empleados, etc.
* Estos datos son utilizados por diversas aplicaciones y sistemas dentro de la organización y deben estar bien gestionados para evitar problemas de calidad de datos y garantizar la toma de decisiones correctas.

**Tipos de datos**:
* No estructurados
* Transaccionales
* Metadatos: datos sobre otros datos
* Jerárquicos
* Maestros: datos críticos de un negocio

**Herramientas de Software** (Ejemplo SAP):
* Modelo de creación de datos y relaciones automatizadas
* Administración de metadatos
* Herramientas de análisis
* Administración de servicios

## Impacto organizacional de los datos deficientes

* **Datos oscuros**: son los datos que se recopilan, procesan y almacenan como parte de las actividades comerciales cotidianas, pero que no la organización no utiliza con ningún otro fin
* **Datos sucios**: en el ámbito de sistemas automatizados, los datos sucios pueden causar a la organización
un daño real, al obligarla a incurrir en un costo económico real causado por las acciones automáticas que dan inicio con datos que no son válidos
