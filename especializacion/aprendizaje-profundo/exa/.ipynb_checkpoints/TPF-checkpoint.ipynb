Procesar en spark utilizando dataframes, las calificaciones y actividades de etiquetado de texto libre de MovieLens*

**MovieLens** es un sistema de recomendación web y una comunidad virtual que recomienda películas a sus usuarios según sus preferencias cinematográficas mediante un filtro colaborativo de las valoraciones y reseñas de los miembros.

### Detalle del dataset

Este conjunto de datos describe la actividad de calificación de 5 estrellas y etiquetado de texto libre de MovieLens (http://movielens.org), un servicio de recomendación de películas.

Los datos se encuentran distribuidos en los archivos `genome-scores.csv`, `genome-tags.csv`, `links.csv`, `movies.csv`, `ratings.csv` y `tags.csv`.

Detalle de los archivos que componen el conjunto de datos:
- **ratings.csv**: cada registro representa la calificación de una película por cada usuario. Las calificaciones se realizan en una escala de 5 estrellas, con incrementos de media estrella (0,5 estrellas - 5,0 estrellas).

	- **Estructura**: ID de usuario, ID de película, calificación,fecha

- **tags.csv**: cada registro representa una etiqueta aplicada a una película por un usuario.

	- **Estructura**: ID de usuario, ID de película, etiqueta, fecha

- **movies.csv**: cada registro representa una película. Los géneros son una lista separada por barras verticales. Cada película puede tener varios géneros asociados.

	- **Estructura**: ID de película, título, géneros

- **links.csv**: cada registro contiene información que permite vincular la película a otras fuentes de datos de películas.

	- **Estructura**: ID de película, ID de imdb, ID de tmdb
		- *movieId* es un identificador de películas utilizado por <https://movielens.org>. Por ejemplo, la película Toy Story tiene el enlace <https://movielens.org/movies/1>.
		- *imdbId* es un identificador de películas utilizado por <http://www.imdb.com>. Por ejemplo, la película Toy Story tiene el enlace <http://www.imdb.com/title/tt0114709/>.
		- *tmdbId* es un identificador de películas utilizado por <https://www.themoviedb.org>. Por ejemplo, la película Toy Story tiene el enlace <https://www.themoviedb.org/movie/862>.

- **genome-scores.csv**: contiene datos de relevancia de las etiquetas de película.

	- **Estructura**: ID de película, ID de etiqueta, relevancia

- **genome-tags.csv**: proporciona las descripciones de las etiquetas para los ID de etiquetas.

	- **Estructura**:  ID de usuario, ID de película, etiqueta, fecha

Los CSVs **genome_scores** y **genome_tags** se consideran más representativos en un solo dataset, se importan juntos.

```python
# Función para importar CSVs en un solo dataset.
def join_csv(estructura):
  df_result = ""

  for dataset in estructura:
    ds = dataset["ds"]
    ds_cols = dataset["ds_cols"]
    ds_join_pks = dataset["ds_join_pks"]

    df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(ds)
    df = df.toDF(*ds_cols)

    if df_result == "":
      df_result = df
    else:
      df_result = df_result.join(df, on=ds_join_pks, how="inner")

  return df_result
```

```python
csv_structure = """{
	"estructura": [
		{
			"ds": "data/genome_scores.csv",
			"ds_cols": [
				"movieId",
				"tagId",
				"relevance"
			],
			"ds_join_pks": [
				"movieId"
			]
		},
		{
			"ds": "data/genome_tags.csv",
			"ds_cols": [
				"tagId",
				"genome_tag"
			],
			"ds_join_pks": [
				"tagId"
			]
		}
	]
}"""
csv_structure = json.loads(csv_structure)
df_genome_scored_tags = join_csv(csv_structure.get("estructura"))
```

### 5. Preguntas de interés - Análisis mediante Gráficos

#### Rating promedio por Tag de Genome y Año entre 1980 y 2010

*Solo se consideran las películas del dataset de movies existentes en el dataset de genome.*

Del dataset de genome_tags:
- Se excluyen etiquetas como: '%imdb top%', 'movielens top%', '%oscar%', 'great acting', '%award%'
- Se agrupan los tags: 'scifi' - 'sci-fi' - 'science fiction'

Los eliminados, no se consideraron relevantes para el análisis.

Para los agrupados, se considera solo el tag de mayor relevancia.

Para ello es necesario:


1.   Agrupar los tags similares y excluir los no representativos.
2.   Filtrar los tags con mayor relevancia por cada película (relevance >= max(relevance)-0.05).
3.   Armar dataset con movieId, title, genome_tag, relevance, rating.
4.   Agruparlos por genome_tag, rating.
5.   Graficar



##### 1. Agrupar los tags similares y excluir los no representativos

```python
# Unificar los tags de sci-fi
df_genome_renamed = df_genome_scored_tags.withColumn(
    "genome_tag",
    when(col("genome_tag").rlike(r'.*sci.*fi.*'), "sci-fi")
    .otherwise(col("genome_tag"))
)

# Eliminar los tags %imdb top%', 'movielens top%'
df_genome_renamed = df_genome_renamed.filter(~col('genome_tag').like('great acting'))
df_genome_renamed = df_genome_renamed.filter(~col('genome_tag').like('%oscar%'))
df_genome_renamed = df_genome_renamed.filter(~col('genome_tag').like('%award%'))
df_genome_renamed = df_genome_renamed.filter(~col('genome_tag').like('imdb top%'))
df_genome_renamed = df_genome_renamed.filter(~col('genome_tag').like('movielens top%'))

# Agrupar los datos, solo dejar los tags más relevantes.
df_genome_renamed = df_genome_renamed.groupBy('movieId', 'genome_tag').agg(max('relevance').alias('relevance'))
```

##### 2. Filtrar los tags con mayor relevancia

```python
df_genome_scored_max = df_genome_renamed.groupBy('movieId').agg(max('relevance').alias('max_relevance'))
```

##### 3. Armar dataset con movieId, title, genome_tag, relevance y rating.

```python
df_genome_filtered = df_genome_renamed.join(df_genome_scored_max, on='movieId').filter(col('relevance') >= (col('max_relevance') - 0.05))
```

##### 4. Agrupar los datos por genome_Tag y rating.

```python
# Join entre df_movie y df_genome_filtered
df_result = df_movie.join(df_genome_filtered, on="movieId", how="inner").select('movieId', 'title', 'year', 'relevance', 'genome_tag')

# Join entre df_result y df_rating
df_result = df_result.join(df_rating, on="movieId", how="inner").select('movieId', 'title', 'year', 'relevance', 'genome_tag', 'rating')

# Agrupar por genome_tag y rating
df_result = df_result.groupBy('year','genome_tag').agg(count(col('movieId')).alias('count'),  round(avg(col('rating')),1).alias('avg_rating'))

df_result = df_result.withColumn("rating", floor(col("avg_rating") * 2) / 2) #Factor = 2

df_result.show(10)
```

##### 5. Graficar

```python
# Convertir el dataset resultante a dataframe de Pandas para graficar
df_pandas = df_result.toPandas()
```

```python
# Calcula el conteo total de películas por cada tag a lo largo de todos los años
tag_counts = df_pandas.groupby('genome_tag')['count'].sum().reset_index()

# Identifica los 10 tags con mayor conteo
top_tags = tag_counts.nlargest(10, 'count')['genome_tag'].tolist()

# Filtra el DataFrame original para incluir solo estos 10 tags
df_top_tags = df_pandas[df_pandas['genome_tag'].isin(top_tags)]

# Fitlrar por periodo 1980 a 2010
df_top_tags = df_top_tags[(df_top_tags['year'] >= 1980) & (df_top_tags['year'] <= 2010)]

fig = px.scatter(
    df_top_tags,
    x='year',
    y='avg_rating',
    size='count',
    color='genome_tag',
    hover_name='genome_tag',
    size_max=50,
    title='Rating vs. Popularidad de Genome Tags por Año',
    labels={'count': 'Conteo de Películas'}
)

fig.show()
```

**Conclusión**

En el gráfico anterior se puede visualizar los tags con mayor rating otorgado por usuarios a lo largo de 30 años (1980 - 2010).

El tamaño de las burbujas depende de la cantidad de reseñas que obtuvo el tag.

A destacar:
- Las burbujas del tag "masterpiece" son muy representativas entre los años 1993 - 1995. Son años en los que se publicaron películas consideradas obras maestras como: Jurassic Park - Philadelphia - Forrest Gump - Apolo 13 - Entre otros.
-  En el periodo 2001 - 2002 destaca el tag "adapted from: book" que puede coincidir con el estreno de las primeras películas de Harry Potter.


#### Evolución de reseñas de peliculas por Tag por Año

Se decide realizar este análisis porque se considera un indicador de que tipo de película (de acuerdo al tag asociado) fue más popular a medida que fueron pasando los años.

Se considera popular aquel tag que tiene mayor cantidad de reseñas.

Para mayor claridad en el gráfico, se seleccionan los primeros 10 tags con mayor cantidad de reseñas.

```python
# Calcula el conteo total de películas por cada tag a lo largo de todos los años
tag_counts = df_pandas.groupby('genome_tag')['count'].sum().reset_index()

# Identifica los 10 tags con mayor conteo
top_tags = tag_counts.nlargest(10, 'count')['genome_tag'].tolist()

# Filtra el DataFrame original para incluir solo estos 10 tags
df_top_tags = df_pandas[df_pandas['genome_tag'].isin(top_tags)]

fig_area = px.area(
    df_top_tags,
    x='year',
    y='count',
    color='genome_tag',
    title='Evolución de reseñas de Películas por Tag por Año (Popularidad)',
    labels={
        'year': 'Año',
        'count': 'Cant. de reseñas',
        'genome_tag': 'Genome Tag'
    },
    height=600
)

fig_area.update_layout(
    xaxis_tickformat='d'
)

fig_area.show()
```

**Conclusión**

En el gráfico anterior (Evolución de reseñas de Películas por Tag por Año) se puede observar un incremento considerable a partir del 1990.

A partir de 1990 comenzó la transición a la filmación digital, con un fuerte crecimiento del cine de ciencia ficción y efectos especiales.

Esto se puede visualizar en el gráfico, siendo que el tag con mayor cantidad de reseñas en ese periodo fue el de "special effects".

#### ¿Que generos reciben las mejores calificaciones? ¿Como evoluciona con el tiempo?

Se excluye del análisis el tag "no genre listed" por no considerarse representativo para este análisis.

```python
# Separar géneros en listas
movies_split = df_movie.withColumn("genre_list", split("genres", "\\|"))

# Obtener lista única de géneros
genres_unique = movies_split.select(explode(col("genre_list")).alias("genre")).distinct()

genre_list = [
    row["genre"]
    for row in genres_unique.collect()
    if row["genre"] != '(no genres listed)'
]

# Crear columnas booleanas para cada género
for genre in genre_list:
    movies_split = movies_split.withColumn(genre, array_contains(col("genre_list"), genre))

# Eliminar columna original de géneros
movies_split = movies_split.drop("genres").drop("genre_list")
```

```python
# Extraer el año en el dataset rating
df_rating = df_rating.withColumn("year", year("timestamp"))
```

```python
# Graficar
plt.figure(figsize=(10,5))

# se agrupa por año y se cuenta películas únicas
movies_by_year = movies_split.groupBy("year").agg(countDistinct("movieId").alias("All_movies")).orderBy("year")

# Ordenar por año y Convertir a pandas para aplicar cumsum
df_base = movies_by_year.toPandas()
df_base["All_movies"] = df_base["All_movies"].cumsum()
```

```python
# Agregar columnas por género
for genre in genre_list:
    genre_movies = movies_split.filter(col(genre) == True)
    genre_count = genre_movies.groupBy("year").agg(countDistinct("movieId").alias(genre)).orderBy("year").toPandas()
    genre_count[genre] = genre_count[genre].cumsum()
    df_base = pd.merge(df_base, genre_count, on="year", how="outer")
```

```python
# Completar valores faltantes
df_base.ffill(inplace=True)
```

```python
# Resetear el índice
df_plot = df_base.reset_index()

# Transformar a formato largo
df_melted = df_plot.melt(id_vars="year",
                         value_vars=[col for col in df_base.columns if col != "All_movies"],
                         var_name="genre",
                         value_name="count")

df_melted = df_melted[df_melted['year']>=1960]

# se crea gráfico de área apilado
fig = px.area(df_melted, x="year", y="count", color="genre",
              title="Evolución acumulada de películas por género",
              labels={"year": "Año", "count": "Cantidad acumulada", "genre": "Género"},
              width=1000, height=500)

fig.update_layout(xaxis_title="Año", yaxis_title="Cantidad acumulada", legend_title="Género")
fig.show()
```

**Conclusión**

Para la confección del gráfico se tuvo extraer la lista de géneros presente en el dataset de peliculas. También, para mayor claridad, se analizaron datos desde 1960.

En el gráfico se puede ver el rating acumulado desde 1960. Las bandas de colores permiten observar el crecimiento en cantidad de personas que califican, pero manteniendo una proporción similar en el ancho de las bandas entre sí (las más calificadas siguen siendo las más calificadas, y las menos calificadas también continúan esa tendencia).

Comedy y Drama se destacan entre todas las demás, con la mayor cantidad de calificaciones, mostrando una tendencia que se sostiene en el tiempo.

#### Agrupar usuarios por ranking y por tag

El objetivo es **detectar** grupos de afinidad para, a futuro, poder agrupar usuarios y hacer recomendaciones.

**1. Expandir el dataset de películas (df_movie) para agregar una fila por cada género listado.**

```python
df_movies_expandido = df_movie.withColumn(
    # 1. Dividir la cadena 'genres' por el delimitador '|' para obtener un Array
    "genre_array",
    split(col("genres"), "\\|")
).withColumn(
    # 2. 'Explotar' el Array, creando una nueva fila para cada elemento
    "single_genre",
    explode(col("genre_array"))
).select(
    # 3. Seleccionar las columnas deseadas y descartar las columnas temporales
    "movieId",
    "title",
    col("single_genre").alias("genre") # Renombrar la columna del género individual
)

df_movies_expandido.show(10, truncate=False)
```

**2. Unir el dataset de películas (df_movies_expandido) con el de rating (df_rating) para obtener el rating, por usuario, por película y género.**

```python
# Unir el df_movies expandido con df_rating
df_usuarios = df_movies_expandido.join(df_rating, on='movieId', how='inner').select('userId', 'title', 'genre', 'rating')

# Eliminar registros asociados a las peliculas sin género
df_usuarios = df_usuarios.filter(col('genre') != '(no genres listed)')

df_usuarios.show(5)
```

**3. Ampliar el dataset para detallar, por usuario, el rating promedio por cada etiqueta.**

```python
# 1. Transformar a formato ancho usando pivot
df_afinidad_ancho = df_usuarios.groupBy("userId").pivot("genre").agg(round(avg(col('rating')),1).alias('rating_promedio'))

# Rellenar los valores nulos (NaN) con 0.0, asumiendo que un rating ausente
# implica que el usuario no ha calificado ese género (o lo ha calificado con un 0).
df_afinidad_ancho = df_afinidad_ancho.na.fill(0.0)

df_afinidad_ancho.show(5)
```

**4. Utilizar K-Means para agrupar a los usuarios en grupos de afinidad de acuerdo a los rating por etiqueta.**

```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.clustering import KMeans

# 1. Identificar las columnas de las características (géneros)
feature_columns = [col for col in df_afinidad_ancho.columns if col != 'userId']

# 2. Crear el Vector Assembler
# Combina todas las columnas de género en una única columna 'features'
assembler = VectorAssembler(
    inputCols=feature_columns,
    outputCol="features"
)

df_kmeans = assembler.transform(df_afinidad_ancho).select("userId", "features")
```

```python
# 3. Entrenar el modelo K-Means
kmeans = KMeans(k=5, seed=1) # k=5: Intenta agrupar usuarios en 5 grupos de afinidad
model = kmeans.fit(df_kmeans)

# 4. Obtener las predicciones (el grupo de afinidad de cada usuario)
df_clustered = model.transform(df_kmeans)

# 5. Unir el cluster al DataFrame original (opcional, para análisis)
df_final_afinidad = df_clustered.select("userId", "prediction").withColumnRenamed("prediction", "affinity_group")

df_final_afinidad.show(5)
```

**5. Graficar los grupos de afinidad para poder visualizar los distintos clusters**

```python
# --- 1. Preparación de Datos (Unir cluster IDs y promedios por género) ---

# Convertir PySpark DataFrames a Pandas
df_afinidad_ancho_pandas = df_afinidad_ancho.toPandas() # Tu DF pivoteado de ratings por género
df_clustered_pandas = df_clustered.select("userId", "prediction").toPandas()
df_clustered_pandas.rename(columns={'prediction': 'affinity_group'}, inplace=True)


# Unir el cluster_id a los ratings por género
df_merged_for_viz = pd.merge(
    df_afinidad_ancho_pandas,
    df_clustered_pandas,
    on='userId',
    how='inner'
)

# Calcular el promedio de rating para cada género, por cada grupo de afinidad
# Excluye 'userId' y 'affinity_group' para promediar solo los géneros
genre_columns = [col for col in df_merged_for_viz.columns if col not in ['userId', 'affinity_group']]

df_cluster_profiles = df_merged_for_viz.groupby('affinity_group')[genre_columns].mean().reset_index()

# Transformar a formato 'largo'
df_cluster_profiles_long = df_cluster_profiles.melt(
    id_vars='affinity_group',
    var_name='Genre',
    value_name='Avg_Rating'
)

# --- 2. Visualización: Gráfico de Barras Agrupadas ---
fig_bar = px.bar(
    df_cluster_profiles_long,
    x='Genre',
    y='Avg_Rating',
    color='affinity_group',
    barmode='group', # Muestra barras agrupadas por género para cada cluster
    title='Perfil Promedio de Rating por Género para Cada Grupo de Afinidad',
    labels={'affinity_group': 'Grupo de Afinidad', 'Avg_Rating': 'Rating Promedio'},
    height=600
)
fig_bar.update_xaxes(tickangle=45)
fig_bar.show()

# --- 3. Visualización: Gráfico Radial ---
# Ideal para comparar visualmente los perfiles si tienes muchos géneros
fig_radar = px.line_polar(
    df_cluster_profiles_long,
    r='Avg_Rating',
    theta='Genre',
    color='affinity_group',
    line_close=True,
    title='Perfil de Afinidad por Género (Gráfico Radial)',
    height=600
)
fig_radar.show()
```

**Conclusión**

Definiendo el modelo K-Means y entrenandolo se pudieron obtener 5 grupos de afinidad que dividen a los usuarios en distintos clústers de acuerdo a los rating que fueron dejando a cada película / género.

De esta manera se puede observar, en los gráficos, las similitudes y diferencias entre cada grupo.

Es especialmente útil para diseñar un sistema de recomendaciones.
