# Apache Kafka
---

* Es una plataforma de streaming **escalable, tolerante a fallos y distribuida con alta disponibilidad** que puede ser utilizada para almacenar y procesar **flujos de datos (data streams)**

## Componentes

* **Kafka cluster**: almacenar data streams (secuencia de mensajes/eventos continuos que producen las aplicaciones). Pueden ser "consumidos" por otras aplicaciones
* **Connect framework (connect API)**: consumir datos y export data streams a sistemas externos (BD, etc)
* **Librería de programación para Streams (Streams API)**: procesar data streams 

## Kafka Cluster

* La arquitectura del cluster provee un servicio de mensajes suscriptor-publicador
* Funciona bajo el esquema **pub/sub**
* **Productores**: escriben y publican mensajes en el cluster
* **Consumidores**: leen mensajes del cluster Kafka
* Los mensajes se almacenan en servidores Kafka llamados **brokers** y luego se los envian a los consumidores
* Un **cluster** está conformado por varios brokers 
* Los brokers tienen capacidad para almacenar, de forma confiable, mensajes en disco
* Los brokes están coordinados mediante **Apache ZooKeeper**
* Los mensajes se organizan en **tópicos** (secuencia de mensajes "logs"). Ej: temperaturas, ventas, pedidos
* Los mensajes se agregan **al final** del tópico
* Los mensajes se definen bajo el esquema **clave-valor**
* Los tópicos se dividen en **particiones**
* Las particiones se replican en diferentes brokers
* Cada mensaje tiene asociado un **offset** (posición del mensaje dentro de la partición)

## Kafka message

* key
* value
* timestamp
* compression type
* optional headers
* partition/offset ID

## Instalación

```sh
$docker run -d --name bro -p 9092:9092 apache/kafka:lastest
$docker ps
$docker container ls
$docker exec -it bro bash
$cd /opt/kafka/bin
$ls
$./kafka-topics.sh --bootstrap-server localhost:9092 --create --topic pruebas
$./kafka-topics.sh --bootstrap-server localhost:9092 --list
$python3.9 -m venv venv
$source venv/bin/activate
$pip install kafka-python

$python consumidor.py
$python productor.py
```
