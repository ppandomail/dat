# Cloud computing
---

* Facilita el acceso y utilización de recursos de cómputo, almacenamiento y aplicaciones
* Analogía con servicios públicos (agua, luz, gas, teléfono)
* Recursos de cómputo se acceden bajo demanda a través de servicios desde cualquier lugar
* HW + Tecnologías de Internet + Administración de Sistemas + Computación Distribuida

## Características

* Autoservicio
* Medición y facturación por uso
* Elasticidad: ilusión de recursos de cómputo infinitos (bajo demanda)
* Personalización: ajustar a las diferentes necesidades de los usuarios
* Simplicidad de uso

## Beneficios

* Reducción de esfuerzo necesario para la instalación y el mantenimiento de los equipos
* Actualización "automática" de recursos
* El costo de inversión está limitado por los servicios consumidos (bajo demanda)
* Reducción de costos asociados al mantenimiento de la infraestructura (energía, refrigeración, seguridad, etc.)
* Reducción de la infraestructura
* Menos personal

### Amazon Elastic Compute Cloud (EC2)

* Permite ejecutar varios "servidores virtuales" Linux o Window bajo demanda
* Brinda acceso de usuario "root o administrador" al SO de cada instancia
* Se factura en función de los recursos consumidos CPU y transferencia de datos

### Amazon Storage

* EBS: servicio de almacenamiento basado en volúmnes (es como utilizar un HD sin formatear)
* S3: almacenamiento de obejtos (archivos y carpetas). Objetos contenidos en depósitos (buckets)

### Amazon CloudWatch

* Monitorización de recursos AWS y aplicaciones.
    * Recopila y visualiza gráficamente métricas de diferentes servicios (EC2, EBS, ELB, RDS, aplicaciones de usuario, etc.)
    * Permite definir alarmas y realizar acciones automatizadas

### Amazon Textract

* Detección y análisis de textos de documentos
    * Texto crudo
    * Layout
    * Formularios
    * Tablas
    * Queries: es un buscador de cosas
    * Firmas
* Cada elemento muestra confianza, poligono de ubicación dentro del documento, etc.

### Amazon Rekognition

* Análisis de imágenes y vídeos
* Identifica objetos, personas, texto, escenas y actividades
* También puede detectar cualquier contenido inapropiado
* También ofrece análisis y reconocimiento facial de alta precisión
* Con las etiquetas personalizadas puede crear un modelo de aprendizaje automático que encuentre los objetos, escenas y conceptos específicos para las necesidades de su negocio
* Muestra confianza de detección de cada elemento que reconoció
