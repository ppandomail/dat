```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

# Cargar el dataset Iris
iris = load_iris()

# Crear DataFrame con todas las variables y la especie
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

# Agregar jitter (ruido pequeño) para evitar superposición
np.random.seed(42)
for feature in iris.feature_names:
    df[feature] += np.random.normal(0, 0.05, size=len(df))

# Lista de nombres de variables
features = iris.feature_names

# Crear figura con 2 filas y 3 columnas
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# Generar los 6 pares únicos de variables con transparencia y jitter
pair_index = 0
for i in range(len(features)):
    for j in range(i + 1, len(features)):
        row = pair_index // 3
        col = pair_index % 3
        sns.scatterplot(data=df, x=features[i], y=features[j],
                        hue='species', ax=axes[row][col],
                        palette='Set1', alpha=0.6)  # Transparencia
        axes[row][col].set_title(f'{features[i]} vs {features[j]}')
        pair_index += 1

# Ajustar diseño
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split

# Cargar el dataset Iris
iris = load_iris()

# Usar solo las dos características de pétalo
X = iris.data[:, 2:]  # petal length y petal width,toma todas las columnas desde la posición 2 en adelante (índice basado en 0).
y = iris.target

# Crear y entrenar el árbol de clasificación
tree_clf = DecisionTreeClassifier(max_depth=2, random_state=42)
tree_clf.fit(X, y)

# Visualizar el árbol
plt.figure(figsize=(8, 6))
plot_tree(tree_clf,
          feature_names=iris.feature_names[2:],  # Solo las dos variables
          class_names=iris.target_names,
          filled=True,
          rounded=True)
plt.title("Árbol de clasificación (Iris)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_moons
# 1. Generar datos
X, y = make_moons(n_samples=100, noise=0.25, random_state=42)

# 2. Entrenar dos árboles: uno sin restricciones y otro con min_samples_leaf=4
tree1 = DecisionTreeClassifier(random_state=42)
tree2 = DecisionTreeClassifier(min_samples_leaf=4, random_state=42)

tree1.fit(X, y)
tree2.fit(X, y)

# 3. Función para graficar fronteras de decisión
def plot_decision_boundary(clf, X, y, ax, title):
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                         np.linspace(y_min, y_max, 200))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.3)
    ax.scatter(X[:, 0], X[:, 1], c=y, edgecolor='k', s=40)
    ax.set_title(title)

# 4. Graficar ambos
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
plot_decision_boundary(tree1, X, y, axes[0], "No restrictions")
plot_decision_boundary(tree2, X, y, axes[1], "min_samples_leaf = 4")
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split

# Cargar el dataset Iris
iris = load_iris()

# Usar todas las características
X = iris.data
y = iris.target

# Crear y entrenar el árbol de clasificación (sin restricciones para ver el árbol completo)
tree_clf = DecisionTreeClassifier(random_state=42)
tree_clf.fit(X, y)

# Visualizar el árbol completo
plt.figure(figsize=(16, 10))  # aumentar tamaño para ver todo
plot_tree(tree_clf,
          feature_names=iris.feature_names,  # todas las variables
          class_names=iris.target_names,
          filled=True,
          rounded=True)
#plt.title("Árbol de clasificación completo (Iris)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree

# 1. Generar datos aleatorios (relación no lineal)
np.random.seed(42)
X = np.sort(5 * np.random.rand(20, 1), axis=0)  # valores entre 0 y 5
y = np.sin(X).ravel()                           # función seno
y[::5] += 3 * (0.5 - np.random.rand(len(y[::5])))  # añadir ruido

# 2. Entrenar árbol de regresión
tree_reg = DecisionTreeRegressor(max_depth=1, random_state=42)
tree_reg.fit(X, y)

# 3. Predicciones para graficar
X_test = np.arange(0.0, 5.0, 0.01).reshape(-1, 1)
y_pred = tree_reg.predict(X_test)

# 4. Graficar con jitter, transparencia y líneas de decisión
plt.figure(figsize=(8, 6))

# Agregar jitter a los puntos
jitter = 0.1 * np.random.randn(len(X))
plt.scatter(X + jitter.reshape(-1, 1), y, color="blue", alpha=0.6, label="Datos reales")

# Línea de regresión
plt.plot(X_test, y_pred, color="red", linewidth=2, label="Árbol de regresión")

# Líneas de decisión (límites de hojas)
thresholds = sorted(set(tree_reg.tree_.threshold[tree_reg.tree_.threshold != -2]))
for thresh in thresholds:
    plt.axvline(x=thresh, color='gray', linestyle='--', linewidth=1)

plt.xlabel("X")
plt.ylabel("y")
#plt.title("Árbol de Regresión con Jitter, Transparencia y Líneas de Decisión")
plt.legend()
plt.grid(True)
plt.show()

# 5. Visualizar estructura del árbol
plt.figure(figsize=(14, 8))
plot_tree(tree_reg, filled=True, rounded=True, feature_names=["X"], fontsize=10)
plt.title("Estructura del Árbol de Regresión")
plt.show()

# 2. Entrenar árbol de regresión
tree_reg = DecisionTreeRegressor(max_depth=2, random_state=42)
tree_reg.fit(X, y)

# 3. Predicciones para graficar
X_test = np.arange(0.0, 5.0, 0.01).reshape(-1, 1)
y_pred = tree_reg.predict(X_test)

# 4. Graficar con jitter, transparencia y líneas de decisión
plt.figure(figsize=(8, 6))

# Agregar jitter a los puntos
jitter = 0.1 * np.random.randn(len(X))
plt.scatter(X + jitter.reshape(-1, 1), y, color="blue", alpha=0.6, label="Datos reales")

# Línea de regresión
plt.plot(X_test, y_pred, color="red", linewidth=2, label="Árbol de regresión")

# Líneas de decisión (límites de hojas)
thresholds = sorted(set(tree_reg.tree_.threshold[tree_reg.tree_.threshold != -2]))
for thresh in thresholds:
    plt.axvline(x=thresh, color='gray', linestyle='--', linewidth=1)

plt.xlabel("X")
plt.ylabel("y")
#plt.title("Árbol de Regresión con Jitter, Transparencia y Líneas de Decisión")
plt.legend()
plt.grid(True)
plt.show()

# 5. Visualizar estructura del árbol
plt.figure(figsize=(14, 8))
plot_tree(tree_reg, filled=True, rounded=True, feature_names=["X"], fontsize=10)
plt.title("Estructura del Árbol de Regresión")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree

# 1. Generar datos aleatorios (relación no lineal)
np.random.seed(42)
X = np.sort(5 * np.random.rand(20, 1), axis=0)  # valores entre 0 y 5
y = np.sin(X).ravel()                           # función seno
y[::5] += 3 * (0.5 - np.random.rand(len(y[::5])))  # añadir ruido

# 2. Entrenar árbol de regresión
tree_reg = DecisionTreeRegressor(max_depth=3, random_state=42)
tree_reg.fit(X, y)

# 3. Predicciones para graficar
X_test = np.arange(0.0, 5.0, 0.01).reshape(-1, 1)
y_pred = tree_reg.predict(X_test)

# 4. Graficar con jitter, transparencia y líneas de decisión
plt.figure(figsize=(8, 6))

# Agregar jitter a los puntos
jitter = 0.1 * np.random.randn(len(X))
plt.scatter(X + jitter.reshape(-1, 1), y, color="blue", alpha=0.6, label="Datos reales")

# Línea de regresión
plt.plot(X_test, y_pred, color="red", linewidth=2, label="Árbol de regresión")

# Líneas de decisión (límites de hojas)
thresholds = sorted(set(tree_reg.tree_.threshold[tree_reg.tree_.threshold != -2]))
for thresh in thresholds:
    plt.axvline(x=thresh, color='gray', linestyle='--', linewidth=1)

plt.xlabel("X")
plt.ylabel("y")
plt.title("Árbol de Regresión max_depth=3")
plt.legend()
plt.grid(True)
plt.show()

# 5. Visualizar estructura del árbol
plt.figure(figsize=(18, 8))
plot_tree(tree_reg, filled=True, rounded=True, feature_names=["X"], fontsize=10)
plt.title("Estructura del Árbol de Regresión")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree

# 1. Generar datos aleatorios (relación no lineal)
np.random.seed(42)
X = np.sort(5 * np.random.rand(20, 1), axis=0)  # valores entre 0 y 5
y = np.sin(X).ravel()                           # función seno
y[::5] += 3 * (0.5 - np.random.rand(len(y[::5])))  # añadir ruido

# 2. Entrenar árbol de regresión
tree_reg = DecisionTreeRegressor( random_state=42)
tree_reg.fit(X, y)

# 3. Predicciones para graficar
X_test = np.arange(0.0, 5.0, 0.01).reshape(-1, 1)
y_pred = tree_reg.predict(X_test)

# 4. Graficar con jitter, transparencia y líneas de decisión
plt.figure(figsize=(8, 6))

# Agregar jitter a los puntos
jitter = 0.1 * np.random.randn(len(X))
plt.scatter(X + jitter.reshape(-1, 1), y, color="blue", alpha=0.6, label="Datos reales")

# Línea de regresión
plt.plot(X_test, y_pred, color="red", linewidth=2, label="Árbol de regresión")

# Líneas de decisión (límites de hojas)
thresholds = sorted(set(tree_reg.tree_.threshold[tree_reg.tree_.threshold != -2]))
for thresh in thresholds:
    plt.axvline(x=thresh, color='gray', linestyle='--', linewidth=1)

plt.xlabel("X")
plt.ylabel("y")
plt.title("Árbol de Regresión sin restricciones")
plt.legend()
plt.grid(True)
plt.show()

# 5. Visualizar estructura del árbol
plt.figure(figsize=(18, 8))
plot_tree(tree_reg, filled=True, rounded=True, feature_names=["X"], fontsize=10)
plt.title("Estructura del Árbol de Regresión")
plt.show()
```
