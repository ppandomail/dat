## Árboles

**Los árboles de decisión son modelos de aprendizaje supervisado que se utilizan para resolver problemas de clasificación y regresión. Empareja cada concepto con su definición correcta:**
* Métrica que mide la impureza de un nodo en clasificación. **Índice Gini**
* Modelo que predice valores numéricos continuos. **Árbol de regresión**
* Parámetro que limita la cantidad de niveles en el árbol. **Max Depth**
* Proceso para reducir la complejidad del árbol y evitar sobreajuste. **Poda (pruning)**
* Modelo que predice clases categóricas. **Árbol de clasificación**

**En árboles de decisión, empareja cada problema con la técnica que lo mitiga:**
* Reducir max_features o aplicar selección de variables. **Complejidad**
* Aplicar poda o limitar profundidad. **Sobreajuste**
* Usar Bootstrapping. **Alta varianza**
* Estimación error Out-of-bag. **Estimación interna del rendimiento del modelo**
* Aplicar técnicas de muestreo o pesos. **Datos desbalanceados**

**Los hiperparámetros en los árboles de decisión controlan la complejidad del modelo y su capacidad de generalización. Empareja cada función con su hiperparámetro:**
* Métrica usada para medir impureza (Gini o entropía). **criterion**
* Número mínimo de muestras en una hoja. **min_samples_leaf**
* Estrategia para elegir la división (best o random). **splitter**
* Número máximo de características consideradas en cada división. **max_features**
* Número mínimo de muestras para dividir un nodo. **min_samples_split**

**El Sobreajuste en  los  árboles de clasificación**
* Ocurre cuando el árbol **memoriza el conjunto de entrenamiento**
* Se puede mitigar con **poda**
* Otra técnica para reducir sobreajuste es **Bagging**

**Sobre los hiperparámetros de un árbol  de clasificación:**
* min_samples_split controla **el número mínimo de muestras para dividir un nodo**
* max_features limita **el número de variables consideradas en cada división**
* criterion define **la métrica de impureza**

**Árboles de clasificación vs regresión:**
* Árbol de clasificación predice **clases categóricas**
* Árbol de regresión predice **valores numéricos**
* En regresión se usa **MSE** como métrica.

**En cuanto a  las Métricas de los árboles de decisión:**
* Índice Gini mide **impureza en clasificación**
* Entropía mide **impureza en clasificación**
* MSE mide **error en regresión**

**Las ventajas del uso de árboles de decisión son las  siguientes:**
* Árboles de decisión son **fáciles de interpretar**
* No requieren **normalización**
* Capturan **relaciones no lineales**

**Un árbol de decisión entrenado con CART:**
* Usa el criterio **Índice Gini** por defecto.
* Si cambiamos a entropía, el árbol será **ligeramente más equilibrado**
* Si aumentamos max_depth sin restricciones, el árbol **puede sobreajustar**

**Cómo  se controla la complejidad de un árbol de clasificación:**
* max_depth limita **la profundidad**
* min_samples_leaf controla **muestras mínimas en hojas**
* Reducir max_features ayuda a **evitar sobreajuste**

**El método de Random Forest:**
* Se basa en **Bagging**
* Reduce **varianza**
* Usa **selección aleatoria de características**

**El índice Gini se utiliza para medir la pureza en árboles de regresión** FALSO

**Limitar la profundidad del árbol reduce el riesgo de sobreajuste.** VERDADERO

**Los árboles de decisión pueden manejar tanto variables categóricas como numéricas.** VERDADERO

**Los árboles de decisión requieren normalización de variables para funcionar correctamente.** FALSO

**El parámetro min_samples_split controla el número mínimo de muestras para dividir un nodo.** VERDADERO

**Los árboles de decisión son modelos lineales.** FALSO

**Random Forest es una técnica basada en Bagging que combina múltiples árboles.** VERDADERO

**Un árbol sin restricciones siempre tendrá mejor rendimiento en datos nuevos.** FALSO

**La poda (pruning) se aplica para reducir la complejidad del árbol** VERDADERO

¿Cómo influye el número de observaciones en la profundidad de un árbol sin restricciones? **A mayor número de observaciones, mayor profundidad**

¿Cómo influye el número de variables en la construcción del árbol? **Más variables ofrecen más opciones de división**

Si dos variables están altamente correlacionadas, ¿qué puede ocurrir? **El árbol elige una y descarta la otra**

¿Qué pasa si agregamos variables irrelevantes? **El árbol puede volverse más complejo. Al principio NO, pero despues SI**

¿Cómo se determina qué variable se elige en la raíz del árbol? **Por la que produce máxima ganancia de información o máxima reducción de impureza**

¿Puede una misma variable aparecer en varios niveles del árbol? **Sí, si sigue siendo útil para separar clases**

¿La impureza de Gini de un nodo suele ser menor o mayor que la de su nodo padre? **MENOR**

Si un árbol de decisión está sobreajustando el conjunto de entrenamiento, ¿es recomendable disminuir la profundidad máxima (max_depth)? **SI**

Si un árbol de decisión está subajustando el conjunto de entrenamiento, ¿es recomendable escalar las características de entrada? **No le afecta el escalado, no es necesario estandarizar o normalizar las variables**

## Análisis de Correspondencia

¿Cuál es el tipo de datos adecuado para aplicar un Análisis de Correspondencias? **Tablas de contingencia con frecuencias no negativas**

En una tabla de contingencia muy grande, el análisis de correspondencias es especialmente útil porque: **Resume la estructura de asociación en pocas dimensiones preservando las distancias Chi-cuadrado**

En el AC, la distancia utilizada para comparar perfiles fila o columna es: **Chi-cuadrado**

En un biplot simétrico, puntos fila y columna cercanos al origen indican **Categorías similares al perfil promedio**

¿Qué representa la "inercia” en el Análisis de Correspondencias? **La variabilidad explicada por los valores propios**

En un mapa de AC, dos categorías columna muy cercanas indican: **Que comparten un perfil similar respecto a las filas**

En un AC simétrico, ¿qué interpretación tiene la proximidad entre un punto fila y un punto columna? **Que presentan una asociación positiva por encima del promedio**

¿Qué tipo de gráfico se utiliza para representar simultáneamente filas y columnas? **Biplot**

¿Qué relación existe entre el AC y el análisis de componentes principales (PCA)? **El AC es el análogo del PCA pero aplicado a tablas de contingencia**

En una tabla de contingencia de gran tamaño (n grande), ¿cuál de las siguientes afirmaciones es correcta respecto al uso del estadístico Chi-cuadrado? **Puede volverse artificialmente grande incluso ante asociaciones muy débiles**
