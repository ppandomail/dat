# ACP (Análisis de Componentes Principales)
---

* Técnica estadística de síntesis de información o reducción de la dimensión (número de variables)
* Es decir, ante un banco de datos con muchas variables, el objetivo será reducirlas a un menor número perdiendo la menor cantidad de información posible
* MUCHAS VARIABLES  ==> POCAS VARIABLES
* Tabla de **datos cuantitativos** $X_{n*p}$
    * columnas: **p variables** cuantitativas
    * filas: **n individuos** 

## Objetivo

* Se quiere obtener una representación en **pocas dimensiones** de los objetos, perdiendo el mínimo de información
* Se quiere obtener (pocas) **variables sintéticas**, basadas en las variables originales y no correlacionadas entre ellas
* Son objetivos equivalentes (y duales)
* **ESPACIO** (nube de datos) ==> **PLANO**   (mayor variabilidad)

## Conceptos claves

* **Transformación de variables**: el ACP convierte un conjunto de variables correlacionadas en un conjunto de variables no correlacionadas (componentes principales), mediante combinaciones lineales. Encuentra relaciones lineales entre las variables. Si no las hay el método no funciona bien
* **Reducción de dimensión**: objetivo es preservar la mayor variabilidad posible del conjunto original, simplificando el problema al reducir el número de variables
* **Técnica descriptiva**: es exploratoria, libre de distribución y no requiere supuestos, por lo que siempre puede aplicarse
* **Maximización de varianza**: busca encontrar las combinaciones lineales que maximizan la varianza de los datos, minimizando la pérdida de información
* **Base matemática**: se parte de una matriz $X \in \mathbb{R} ^{n+p}$, donde se registran $p$ variables sobre $n$ individuos
    * si $n < p$ se podrá encontrar n componentes principales
    * si $n > p$ se podrá encontrar p componentes principales 

## Aspectos claves

* Todas las variables tienen el mismo rol: **no hay independientes ni dependientes**
* El objetivo es **reducir la dimensión del problema**, descartando información redundante
* Permite visualizar datos multidimensionales
* Se explora la existencia de variables latentes
* Su aplicación tiene sentido solo si las variables están fuertemente correlacionadas
* Se recomienda usar la **matriz de correlaciones** (no la de covarianzas) para evitar distorsiones por unidades de medida
* Es necesario **definir un criterio de finalización** para seleccionar el número de componentes
* Los análisis confirmatorios ayudan a evaluar la estabilidad de los componentes y detectar observaciones atípicas (en gral. si se sacan y se vuelve a correr el análisis de ACP ajuste mucho mejor y aumenta la varianza explicada)
* Puede ser útil para identificar anomalías en los datos

## Análisis de componentes principales 

* Si los puntos se distribuyen de la manera que el Gráfico 1, se busca el subespacio donde se maximice la varianza 
* Encontrar la recta y luego la ortogonal (perpendicular) que cumpla con lo mismo, es decir, se maximice la varianza

## Análisis de la matriz de correlaciones

* Un ACP tiene sentido si existen **altas correlaciones lineales entre las variables**, ya que esto es indicativo de que existe información redundante y, por tanto, pocos factores explicarán gran parte de la variabilidad total

## Selección de factores

* La elección de los factores se realiza de tal forma que el primero recoja la mayor proporción posible de la variabilidad original
* El segundo factor debe recoger la máxima variabilidad posible no recogida por el primero, y así sucesivamente
* Del total de factores se elegirán aquellos que recojan el porcentaje de variabilidad que se considere suficiente
* A estos se les denominará componentes principales

* Se tiene una nube de puntos y se ubica una eje (PC1) donde se tiene la mayor variabilidad de los datos
* Luego otro eje (PC2) que será perpendicar a PC1, donde va a tener la otra dirección de varianza
* PC1, PC2: son los componentes principales
* Se genera 4 cuadrantes alrededor del centro de masa $(\bar x_1; \bar x_2)$
* **Cov: covarianza muestral** (mide la relación lineal entre dos variables). Hace que los valores estén centrados en la media
    * n: número total de observaciones 
* **r: coeficiente de correlación lineal de Pearson** (versión normalizada de la covarianza, entre [-1, 1]
    * S: desviación estándar de la variable x
    * S2: desviación estándar de la variable y 

## Rotación a las componentes

* **Y1: CP1** y **Y2: CP2**
* $y_{i1}$ y $y_{i2}$ son los **autovectores**
* variables de 1 a p
* individuos de 1 a n
* $v_{1j}$ a $v_{pj}$ es el autovector
* La dirección del autovector tiene que ser **MAXIMA VARIANZA**
    * Si hay 10 variables hay 10 dimensiones
    * Buscar la dirección donde haya máxima varianza
* La varianza en ese nuevo espacio, es que la varianza de cada eje nuevo va a ser el valor del autovector
* La suma de la diagonal principal (TRAZA) de los autovalores = suma de las varianzas de cada variable

## Ejemplo

* **Matriz de correlación/coeficientes**

|    | v1    | v2   | v3   |
| -- | --    | --   | --   |
| v1 | 1.00  |      |      |
| v2 | -0.39 | 1.00 |      |
| v3 | -0.01 | 0.40 | 1.00 |

* La suma de la diagonal principal es igual a la cantidad de variables (p = 3).
* Con **3 variables** tengo **3 autovectores** y **3 autovalores**
* La varianza depende de cada variable Var(v1) = 0.33, Var(v2) = 0.33, Var(v3) = 0.33
* todas las variables tienen una repartija igual de la varianza

---

* **Matriz de correlación/Probabilidades**

|    | v1      | v2      | v3   |
| -- | --      | --      | --   |
| v1 |         |         |      |
| v2 | <0.0001 |         |      |
| v3 | 0.8780  | <0.0001 |      |

---

* **Autovalores**: porción de varianza que explica cada componente

| λ  | Valor | Proporción | Prop Acum |
| -- | --    | --         | --        |
| λ1 | 1.57  | 0.52       | 0.52      |
| λ2 | 0.99  | 0.33       | 0.85      |
| λ3 | 0.45  | 0.15       | 1.00      |

* $\Sigma_{Valor} = 3$
* Los valores se ponen de orden decreciente
* Proporción: 1.57 / 3 = 0.52. $\Sigma_{Proporción} = 1$. Es la proporción de la varianza
* La Var(CP1) = λ1, Var(CP2) = λ2, ... se va a tener un nuevo espacio donde PC1 explica el el 52% de la variabilidad de los datos (mas de la mitad de la varianza se encuentra concentrada alli)
* 0.52 significa que explica el 52% de la variabilidad total de los datos
* La repartija de la varianza depende del autovalor; cuanto mas grande el valor mas concentración en el primer eje de la variabilidad
* Si tengo 3 variables y tengo un espacio de 3 variables no tendría ninguna pérdida de variablidad

---

* **Autovectores** que marcan la dirección, que van a ser perpendiculares entre si y van a ser las direcciones de máxima variabilidad

|    | e1    | e2      |
| -- | --    | --      |
| v1 | -0.50 | 0.71    |
| v2 |  0.70 | 5.3E-04 |
| v3 |  0.51 | 0.70    |

* e1: espacio con el 52%
* e2: espacio con el 33%
* no hay e3, renuncio al 15%
* e1 + e2 = 85% de la variabilidad total de los datos

---

* **Correlaciones con las variables originales**

|    | CP1   | CP2     |
| -- | --    | --      |
| v1 | -0.62 | 0.71    |
| v2 |  0.88 | 5.3E-04 |
| v3 |  0.64 | 0.69    |

* cargas factoriales, de como se proyectan las variables en los ejes
$$λ1 = (-0.62)^2 + (0.88)^2 + (0.64)^2 = 1.57$$
$$CP1 = -0.62 * X_1 + 0.88 * X_2 + 0.64 * X_3$$
$$CP2 = 0.71 * X_1 + 0 * X_2 + 0.60 * X_3$$

* La variable **lag13** tienen 0.88 en CP1 y 0 en CP2. La variabilidad en el eje 1 es muy importante y no tiene importancia en el eje vertical CP2
* Se identifican individuos con **alto lag13** y con **bajo lag13**
* Con los 3 ejes es el 100% de la varianza, con 2 ejes es el 85% de la varianza
* **Cos2**: es el cuadrado de la dimensión. Da una idea del coeficiente que va a multiplicar las variables en cada una de las dimensiones
* **Comunalidad (Coseno2)**: que tanto está explicando la variable con el $CP1^2 + CP2^2$. Casi el 89% de la varianza de la variable es bastante. Cuanto de la variable está explicado en esos dos ejes
* Se observa con la comunalidad que hay alta explicación de las variables individuales, y alta explicación de todo el sistema

## Observaciones

* Todas las variables juegan el mismo papel, es decir, el conjunto no se divide en variables dependientes e independientes como en el caso de la regresión
* El objetivo de ACP es reducir la dimensión del problema. Eliminar variables redundantes, o bien información redundante
* Permite detectar la presencia de variables latentes (las componentes!)
* Si las variables originales no están fuertemente correlacionadas, no tiene sentido buscar componentes principales
* Las unidades de medición pueden influir en la variabilidad de las componentes principales, luego para seleccionar el método conviene considerar la matriz de correlaciones y no la de covarianzas
* Debemos establecer un criterio para seleccionar la cantidad de componentes principales a seleccionar

## Ventajas

* Reducir el número de variables de los datos eliminando las que son ruido, lo que también puede reducir el sobreajuste
* Mejorar el rendimiento de los algoritmos posteriores centrándose solo en las características relevantes
* Mejorar la visualización de los datos para comprenderlos mejor
* Elimina colinealidad

## Supuestos

* Existe una combinación lineal entre variables
* Supone que los componentes principales que tienen mayor varianza son más importantes que los que no la tienen, y considera que incluyen ruido en los datos
* Los valores atípicos en los datos incluyen errores experimentales
* El conjunto de datos del ACP ofrece una gran representación de los datos originales

## Limitaciones

* En primer lugar, el PCA solo funciona con variables numéricas
* En segundo lugar, no funciona bien cuando las variables no están fuertemente correlacionadas (correlación lineal débil)
* Además, el PCA es sensible a la escala de las características y el resultado se ve afectado por los valores atípicos

**¿Cuándo debes utilizar el análisis PCA?**
* Se recomienda utilizar el PCA cuando se trate con variables con fuerte correlación lineal

**Test de esfericidad de Barlett**
* H0: no hay direcciones de máxima variabilidad (hay esfericidad)
* Si tengo una pelota bien esferica, el método no sirve

**Los puntos que están cerca del (0,0) no aportan**

## Cuantas componentes me quedo

* **Criterio 1**: porcentaje de variabilidad mínimo que se desea explicar (% de varianza explicada)
* **Criterio 2 (Kaiser)**: retener las m primeras componentes tales que su autovalor resulte igual o mayor a 1 (o mayor a 0.7-0.8)
* **Criterio 3 (Bastón roto)**: en un gráfico de sedimentación, ver la cantidad de componentes hasta que se estabilice el gráfico
* **Criterio 4**: se seleccionan aquellas componentes cuyo autovalor excede de la media de los autovalores

### Criterios para decidir el número de componentes principales

* **Porcentaje de varianza explicada**: algunos el 70%, otros el 80%, otros 90% depende de las cracteristicas específicas del estudio que se está realizando
* **Criterio de Kaiser**: quedarse con las componentes principales relacionadas a los autovalores mayores que 1 o muy próximos a 1
* **Criterio del bastón roto**: graficar la varianza explicada en función del número de componentes y se busca el luegar donde se quiebra y se estabiliza la varianza
* **Prueba de esfericidad**: es un test de hipótesis que se utiliza muy poco

## Procedimiento

1. Normalización de los datos
2. Explorar Matriz de covarianza o correlaciones: si hay correlaciones muy bajas no tiene sentido ACP. Tiene que ser r > 0.5 o r < -0.5
3. Buscar Autovectores y Autovalores
4. Selección de los componentes principales
5. Transformación de los datos en nuevo espacio dimensional

## Conclusiones

* Sabemos que los conjuntos de datos grandes son cada vez más frecuentes y, a menudo, difíciles de comprender. El Análisis de Componentes Principales reduce la dimensionalidad de estos conjuntos de datos, a la vez que aumenta la
interpretabilidad y minimiza la pérdida de información. Maximiza la varianza gradualmente al introducir nuevas variables no correlacionadas
* PCA se utiliza principalmente para reducir la dimensionalidad en muchas aplicaciones de IA, como visión artificial, compresión de imágenes, etc
* Si los datos tienen una gran dimensión, también pueden utilizarse para descubrir patrones ocultos
* El PCA suele buscar la superficie de menor dimensión sobre la que proyectar los datos de mayor dimensión. Por eso, el PCA resulta beneficioso y práctico.
