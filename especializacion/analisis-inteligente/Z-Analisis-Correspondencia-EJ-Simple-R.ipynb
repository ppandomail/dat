# Ejemplo R - Análisis de Correspondencia

```r
library(readxl)
```

```r
m <- read_excel("datos/ejemplosimple.xlsx")
```

```r
M <- as.matrix(m)
```

```r
# Guarda la matriz que caracteriza las filas
Mf <- M[,2:3]
```

```r
# Guarda la matriz que caracteriza las columnas
Mc <- M[,4:6]
```

```r
# Arma la tabla de contingencia
F <- t(Mf) %*% Mc
```

```r
# Calcula el vector totales fila
totalf <- F %*% rep(1, 3)
Df <- diag(as.vector(totalf))
```

```r
# Arma la matriz diagonal con las frecuencias de las filas
R  <- solve(Df) %*% F # Calcula la matriz R
round(R,3)            # Exhibe el resultado con 3 decimales
```
