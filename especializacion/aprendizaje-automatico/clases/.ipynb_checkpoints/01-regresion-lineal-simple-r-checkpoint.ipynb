# Regresión Lineal Simple - R

* Estudiar si existe un comportamiento lineal entre dos variables
numéricas
* Ver si una de las variables (variable independiente) nos ayuda a predecir o explicar el comportamiento de otra variable (dependiente)
* Ejemplo: Estudiar cómo cambia el peso de una persona de acuerdo a su edad
* ¿Podemos afirmar, con un cierto nivel de significación, que existe evidencia de que el peso de una persona tiende a aumentar con la edad?
* ¿Podemos predecir, aproximadamente, la cantidad de grasa en sangre según la edad de la persona? ¿Qué grado de confianza tiene esa predicción?

* Objetivo: Encontrar la recta que mejor ajusta a los datos
* Hay que definir un criterio para elegir la recta que mejor ajusta a los datos --> y = β0 + β1x
* Predicho: punto sobre la recta --> (x, ŷ)
* Residuo: distancia entre el punto y la recta --> r = y - ŷ

* Encontrar los parámetros β0 y β1
    * Matemáticos: Método de Mínimos Cuadrados (MC)
    * Estadístico: Modelo de Regresión Lineal (MRL)

* Covarianza entre dos VA --> Cov(X, Y) = E[(X − E(X))(Y − E(Y))]
    *  Cov(X, Y) > 0: ambas variables tienden a aumentar juntas (relación positiva)
    *  Cov(X, Y) < 0: cuando una variable aumenta, la otra tiende a disminuir (relación negativa)
    *  Cov(X, Y) ≈ 0: no hay relación lineal clara

* Coeficiente de correlación lineal: ρxy = Cov(X, Y) / (sqrt(Var(X)) * sqrt(Var(Y)))
* Toma valores entre [-1, 1]

## Modelo de regresión lineal simple

* Estudia la relación entre dos variables aleatorias:
    * X = variable predictora, regresora o explicativa
    * Y = variable dependiente o respuesta

* Permita:
    * Analizar si existe una relación lineal entre ambas variables
    * Cuantificar esta relación mediante un estudio del modelo
    * Predecir el valor de Y para un valor dado de X = x

* Forma: Yi = β0 + β1Xi + εi
  * β0: ordenada al origen o intercept, indica el valor medio de Y cuando X = 0
  * β1: pendiente, indica la variación media de la variable respuesta cuando X varía en una unidad
  * εi: errores --> εi ∼ N(0, σ^2)
* Se desconocen los parámetros: β0, β1 y σ. Hay que estimarlos
* Método de Máxima Verosimilitud (MV) para estimar los parámetros: ^β0 MV, ^β1 MV y ^σ^2MV
* Encontrar el valor del parámetro que maximiza la función de verosimilitud, es decir, la probabilidad (o densidad) de observar los datos muestrales, dados esos parámetros

## 1.2 Relación entre variables

Se puede explorar cómo se relacionan las variables entre sí utilizando un gráfico de dispersión (scatter plot) que muestra los puntos correspondientes a cada par de variables. Para generar estos gráficos, utilizamos las librerías ggplot2 y GGally, que deben estar previamente instaladas y cargadas en el entorno de trabajo

Para ver la estructura del archivo utilizamos la funcion str

```r
str(grasas)
```

Para realizar el gráfico de dispersión utilizamos la función ggpairs

```r
# Gráfico de dispersión (para explorar cómo se relacionan las variables entre sí)
ggpairs(grasas,
        columns = 1:3,
        mapping = aes(alpha = 0.9),
        upper = list(continuous = wrap("points", color = "darkred")),
        lower = list(continuous = wrap("points", color = "blue")),
        diag  = list(continuous = wrap("densityDiag", fill = "lightblue")))
```

El gráfico sugiere la presencia de una relación lineal entre las variables edad y grasas. En cambio, no se observa una asociación evidente entre los otros pares de variables

```r
# scatter plot: gráfico de puntos
ggplot(data = grasas, aes(x = edad, y = grasas)) + 
    geom_point(size = 2)
```

Planteamos un modelo de regresión lineal

```r
# Primero va la variable respuesta y luego la explicativa
modelo <- lm(formula = grasas ~ edad, data = grasas)
modelo
```

```r
summary(modelo)
```

Los parámetros estimados de la recta que modela la cantidad de grasas en sangre en función de la edad se encuentran en la columna “Estimate” de la tabla “Coefficients” de la salida mostrada anteriormente. Por lo tanto, en este caso, la ecuación de la recta que mejor ajusta a este conjunto de datos es:

$$ 𝑦 = 102.575 + 5.321𝑥$$

# 2. Interpretación de los parámetros del modelo

* El intercept u ordenada al origen, **102.575**, representa la cantidad estimada de grasas en sangre cuando la edad es igual a cero. Aunque este valor no es interpretable directamente en el contexto del problema, es necesario para definir la ecuación de la recta de regresión

* El coeficiente de la variable edad (5.321) indica que, en promedio, un incremento de una unidad en el edad se asocia con un aumento de 5.321 unidades en la cantidad de grasas en sangre

# 3. Extracción de los coeficientes del modelo

# 8. Diagnóstico de la regresión

* Son técnicas que permiten validar el cumplimiento de los supuestos del modelo planteado
* Se basan en el análisis de los residuos
* Los residuos son observables ri = Yi − ^Yi
* Los residuos no son independientes, están correlacionados
* Gráficos de residuos (ri) vs la covariable (Xi)
* ¿Cómo debería verse el gráfico de residuos vs. la covariable (o los predichos) si el modelo es adecuado?
* Si el modelo es correcto, el gráfico debería lucir como una nube de puntos sin estructura, ubicada alrededor del eje horizontal

## 8.1 Análisis de los residuos

### 8.1.1 Plot de residuos estandarizados vs predichos

```r
# residuos estandarizados tienen varianza 1
res.pred.df <- data.frame(x = modelo$fitted.values, y = rstandard(modelo))
ggplot(data = res.pred.df, aes(x, y)) +
  geom_point(size = 1) +
  geom_hline(yintercept = 0, color = "red") +
  theme_minimal() +
  labs(title = "Residuos estandarizados vs. predichos",
       x = "valores predichos",
       y = "residuos estandarizados")
```

### 8.1.2 Plot de residuos estandarizados vs variable regresora

```r
res.reg.df <- data.frame(x = grasas$edad, y = rstandard(modelo))

ggplot(data = res.reg.df, aes(x, y)) +
  geom_point(size = 1) +
  geom_hline(yintercept = 0, color = "red")
```

Se observan puntos aleatorios sin ninguna estructura (patrón). Quiere decir que las variables explicativas están en el modelo
* Se consideran a los errores como homocedásticos (varianza constante)
* Se consideran que los errores son independientes

### 8.1.3 Plot de residuos estandarizados vs variable regresora

```r
# qqplot residuos estandarizados (testeo que garantiza normalidad)
ggplot(data = data.frame(residuos = rstandard(modelo)), aes(sample = residuos)) +
  stat_qq(color = "dodgerblue") +
  stat_qq_line(color = "red", linetype = "dashed") +
  theme_minimal() +
  labs(title = "Q-Q Plot de residuos estandarizados",
       x = "Cuantiles teóricos",
       y = "Cuantiles muestrales")
```

# 9. Calidad del ajuste del modelo

* Descomposición de la suma de cuadrados

* Yi − ¯Y = Yi − ^Yi + ^Yi − ¯Y
* Desviación total:                                   Yi − ¯Y
* Desvío de los predichos respecto de la media:      ^Yi − ¯Y
* Desvío alrededor de la recta de regresión ajustada: Yi − ^Yi

* **Coeficiente de determinación R^2**: indica qué parte de la variabilidad total de la variable dependiente Y es explicada por la variable independiente. Por lo tanto, sirve como una medida del poder predictivo del modelo. Nos habla sobre la calidad de ajuste del modelo. Queremos que sea próximo a 1 (buen modelo, arriba de 0.8 relación fuerte. Entre 0.6 y 0.7 la relación es débil)

```r
summary(modelo)
```

* Multiple R-squared:  0.7012  **es R^2**
* El valor 0.7012 implica una relación lineal moderadamente fuerte entre la edad de la persona y la cantidad de grasas en sangre
* Este valor indica que el modelo explica un 70.12 % de la variabilidad observada
* Por lo tanto, un 29.88 % de la variabilidad observada no queda explicada por el modelo

# 10. Inferencia en el modelo de regresión simple

## 10.1 Intervalo de confianza para ordenada al origen y pendiente

```r
# Intervalos de confianza para los parámetros
confint(modelo) # nivel de confianza por default es 0.95
```

* Este intervalo [41.26, 163.88] tiene una confianza del 95% de contener al verdadero valor de β0
* Si se repitiera 100 veces el experimento, en el 95% de esas veces va a caer en ese intervalo 

```r
# Cambiando el nivel de confianza
confint(modelo, level = 0.90)
```

## 10.2 Intervalo de confianza para sigma

```r
# Intervalo de confianza para 𝜎^2
# El estimador de la desviación típica de los errores, 𝜎, figura como Residual standard error 
# y su valor en el ejemplo es 43.5. Este valor se puede obtener con la instrucción sigma(regresion)
# n-2 son los grados de libertad
alfa <- 0.05
n <- dim(grasas)[1]
# Límite Superior
LSsigma <- (n-2) * sigma(modelo)^2 / qchisq(1-alfa / 2, n-2)
# Límite Inferior
LIsigma <- (n-2) * sigma(modelo)^2 / qchisq(alfa / 2, n-2)
LSsigma
LIsigma
```

## 10.3 Test de Hipótesis para β0 y β1

* La columna t value muestra el valor del estadístico T, que es el cociente entre cada estimador y su error estándar. Estos valores son utilizados para realizar los contrastes de hipótesis
    * H0: β0 = 0   vs    H1:  β0 != 0
    * H0: β1 = 0   vs    H1:  β1 != 0

* El t-value de β1 = 5.3207 / 0.7243 = 7.346 (lo dá el summary)
* El p-value es la probabilidad que el módulo del estadístico sea mayor que el módulo t-calculado (t-value). Y compara esa probabilidad con el nivel de significación
* P(T > 7.346) + P (T < -7.346)  
* Los p-valores correspondientes a estos contrastes se encuentran en la columna Pr(>|t|) 

```r
2 * pt(7.346, 23, lower.tail=FALSE)
```

```r
pvalores <- summary(modelo)$coefficients[, 4]
names(pvalores)
pvalores
```

En este caso, como los p-valores son muy pequeños se rechazan ambas hipótesis para los niveles de significación clásicos

**Conclusión**: para este problema, ambos parámetros son significativamente != 0. Es decir, hay suficiente evidencia para afirmar que ambos parámetros son != 0

# 13. Outlier (valor atípico)

* Es una observación que tiene un valor alto de residuo
* Puede influir en distintos aspectos del modelo de regresión lineal

```r
res_df <- data.frame(x=modelo$fitted.values, y=rstandard(modelo))

ggplot(data=res_df, aes(x,y)) +
  geom_point(size=2) +
  geom_hline(yintercept = 0, color="red") +
  geom_hline(yintercept = 2, color="green", linetype="dashed") +
  geom_hline(yintercept = -2, color="green") +
  labs(title = "Residuos estandarizados",
       x = expression(hat(y)[i]), 
       y = expression(hat(r)[is])) +
  theme_minimal()
```

* Tipos de outliers en regresión

| Tipo | Donde? | Qué puede afectar? |
| -- | -- | -- |
| en Y (residuo grande) | valor de x normal, y muy distinto | aumenta el error, pero no cambia la recta mucho |
| en X                  | valor de x extremo                | puede tirar la recta hacia sí mismo |
| influyente            | extremado en x y alejado de la recta | cambia la pendiente o intercepto del modelo |

* En qué influye un outlier?
    * **coeficientes**: pueden ser distorsionados si el punto es influyente
    * **residuos**: aumenta el error cuadrático medio
    * **R^2**: puede disminuir si aumenta la variabilidad no explicada
    * **Inferencia**: afecta a los valores p, IC y supuestos (normalidad, homocedasticidad)

## Análisis de puntos influyentes

```r
# Leverage: lejanía del predictor xi respecto al centro
leverage <- hatvalues(modelo)
leverage
```

```r
# Residuos estandarizados
residuos_std <- rstandard(modelo)
residuos_std
```

```r
# Distancia de Cook: influencia global del punto sobre el modelo
# Cuanto cambia el modelo al eliminar el punto
cooks <- cooks.distance(modelo)
cooks
```

```r
# Umbrales típicos
n <- nrow(grasas)

p <- length(coef(modelo)) 
leverage_umbral <- 2 * p / n
cooks_umbral <- qf(0.5,p,n-p)
```

```r
# Tabla con diagnóstico
diag <- data.frame(
  obs = 1:n,
  edad = grasas$edad,
  grasas = grasas$grasas,
  leverage = leverage,
  rstandard = residuos_std,
  cooks = cooks
)
```

```r
# Puntos con Leverage alto
diag[diag$leverage > leverage_umbral, ]
```

```r
# Puntos con distancia de Cook alta
diag[diag$cooks > cooks_umbral, ]
```

```r
leverage_umbral<- 2 * p / n
cooks_umbral <- qf(0.5,p,n-p)

# Agregamos el número de observación al data frame

datos <- data.frame(obs = 1:dim(grasas)[1], grasas)

ggplot(datos, aes(x = leverage, y = cooks, label = ifelse(leverage > leverage_umbral | cooks > cooks_umbral, obs, ""))) +
  geom_point(color = "steelblue", size = 2) +
  geom_hline(yintercept = cooks_umbral, linetype = "dashed", color = "darkgreen") +
  geom_vline(xintercept = leverage_umbral, linetype = "dashed", color = "red") +
  geom_text(nudge_x = 0.001, nudge_y = 0.001, size = 3) +
  labs(
    title = "Leverage vs Cook's Distance (modelo: grasas ~ edad)",
    x = "Leverage",
    y = "Cook's Distance"
  ) +
  theme_minimal()
```
