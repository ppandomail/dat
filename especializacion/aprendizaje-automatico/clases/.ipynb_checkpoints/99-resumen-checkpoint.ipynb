## Factores de Riesgo

### Tabla de Factores de Riesgo

```
         E
      SI  NO
F SI  SS  SN
  NO  NS  NN
```

* E: Efecto (variable objetivo). Ej: bajo peso 
* F: Factor (variable independiente). Ej: tabaquismo madre
* Prevalencia de bajo peso = C_SI / ALL
* **F1**: es la media armónica entre precisión y sensibilidad. Se busca un F1 grande  (cercano a 1)

|    |                     |    |    |    |
| -- | --                  | -- | -- | -- |
| RR | Riesgo Relativo     | Compara la incidencia entre los que tienen factor de riesgo y quienes no lo tienen | RR = (SS / F_SI) / (NS / F_NO) | RR = 1 no hay riesgo |
| OR | Odds Ratio (chance) | Compara chances de los que tienen factor vs los que no                             | OR = chance si fuma / chance no fuma = (SS / SN) / (NS / NN) | OR = 1 => F no representa riesgo; OR > 1 => F es dañino; OR < 1 => F es protección |

## Regresión Logística

### Concepto

* Modelar la relación para una variable respuesta dicotómica con regresión
* Ejemplo:
$$Y = 1 \text{ si presenta incidente cardíaco}$$
$$Y = 0 \text{ si no presenta incidente cardíaco}$$
* La Prob de presentarse el efecto para un determinado valor de x (edad) es
* Se propone modelar esta probabilidad, es decir la $P(Y=1)$ según sea el valor de $x$
* **Función Logística**
$$\pi(x) = P(Y = 1/x) = \dfrac{e^{\beta_0 + \beta_1x}}{1 + e^{\beta_0 + \beta_1x}}$$

* $\pi(x)$: es monótona (creciente o decreciente)

### Regresión Logística múltiple

* Modelar las probabilidades condicionales mediante una función logística
$$PI(x) = P(Y=1 | x_1, x_2, ..., x_p)$$
* Las variables predictoras pueden ser continuas o categóricas
* Los coeficientes son interpretables

### Desvianza

* Se utiliza para:
	* **testear la significancia del modelo**: vale la pena modelar $Y$ a partir de $x_1, ... x_p$? Se compara dev(modelo) con la devianza del modelo nulo (con sólo el término constante)
        * H0: modelo solo con $\beta_0$ (modelo null)
        * H1: modelo completo
        * pchisq(diferencia entre devianza nula y devianza residual). Rechazar la hipótesis nula (modelo es signficativo)
	* **testear la significatividad de una variable**: vale la pena agregar esa variable a las que ya puse? Se compara el modelo con y sin esta, utilizando las devianzas correspondientes
	* **comparar modelos anidados en general**

### Regresión Logística para clasificación

* El modelo logístico predice la probabilidad de que se presente el evento $Y=1$ en cada nivel de covariables
* Manera comun de clasificar: si prob > 0.5 entonces caso con clase $Y=1$

### Matriz de confusión

* Son métricas de evaluación para analizar la capacidad de predicción y clasificación de clases del modelo construido
* Permite comparar el resultado obtenido a partir del modelo predictivo construido con los resultados de los datos de prueba del modelo

```
                                Real
                     1 (positivo)  0 (negativo)
         1 (positivo)      VP           FP
Predicho 0 (negativo)      FN           VN
```

* VP: #casos clasificados como 1 y son de clase 1
* VN: #casos clasificados como 0 y son de clase 0
* VP + VN: decisiones correctas
* FN + FP: representan los errores (la confusión) entre las clases (F: Falso)
* total = VP + VN + FN + FP

| Métrica                  | Fórmula           | Observaciones                                                                            |
| --                       | --                | --                                                                                       |
| **Accuracy (exactitud)** | (VP + VN) / total | evalúa la capacidad de generalización del modelo para predecir y clasificar nuevos casos |
|                          |                   | ALTA: clasificaciones correctas >= 70% casos                                             |
|                          |                   | BAJA: clasificaciones correctas < 70% casos (modelo poco confiable)                      |
| **Sensitividad**         | VP / (VP + FN)    | Es la capacidad del modelo para identificar correctamente los casos positivos reales     |
| **Especificidad**        | VN / (FP + VN)    | Es la capacidad del modelo para identificar correctamente los casos negativos reales     |
| **Precisión**            | VP / (VP + FP)    | evalúa la efectividad del modelo para clasificar casos a una clase particular            |
|                          |                   | ALTA: modelo efectivo para predecir y clasificar nuevos casos                            |
|                          |                   | MEDIA: modelo inestable. Posible confusión en clasificación y predicción                 |
|                          |                   | BAJA: modelo confunde las clases. Modelo poco efectivo                                   |

* **Probabilidad de mal clasificados** = 1 - accuracy
* La matriz depende fuertemente de las prob predichas
* S y E dependen de la "mezcla" de casos
* Muchas veces se quiere mejor Sensitividad que Especificidad
* Frecuente usar curva ROC para elegir modelo
* El área bajo la curva ROC mide la capacidad predictiva del modelo

## Naive Bayes

### Concepto

* Es un algoritmo de **aprendizaje supervisado** basado en el teorema de **Bayes**
* Se utiliza para tareas de clasificación, como la filtración de spam o la clasificación de textos. Es multiclase
* Asume que todos los atributos son independientes entre sí, lo que simplifica enormemente los cálculos de probabilidad, a pesar de que esta suposición raramente se cumpla en el mundo real
* Si se quiere calcular la probabilidad de un evento dado otro, se puede hacer en base a esta fórmula:

$$𝑃(𝐴/𝐵) = \dfrac{𝑃(𝐵/𝐴) 𝑃(𝐴)}{𝑃(𝐵)}$$

```mermaid
graph TD
    C(C) --> A1(A1)
    C --> A2(A2)
    C --> A3(..)
    C --> An(An)
```

* C: variable de clasificación
* $A_i$: atributos predictores
* Dado un registro con atributos A1,...,An el objetivo es predecir la clase C 

* **¿Cómo predecir C a partir de los atributos?**
  * Buscar el valor de C que maximice las probabilidades a posteriori $P(A_1, A_2,..., A_n / C) * P(C)$

* Aprendiza paramétrico: determinar probabilidades a priori de cada clase y las probabilidades condicionales
    * P(atrib obs/clase1) = #atr1/#clase1 * ... * #atrn/#clase1
    * P(atrib obs/clase2) = #atr1/#clase2 * ... * #atrn/#clase2
    * PC1 = P(atrib obs/clase1) P(clase1) = P(atrib obs/clase1) * #clase1/#poblacion
    * PC2 = P(atrib obs/clase2) P(clase2) = P(atrib obs/clase2) * #clase2/#poblacion
    * Si PC1 > PC2 => se lo clasifica como C1
* Es decir, se mira las condiciones de los atributos observados y pondera por la representatividad de las clases sobre las que quiero clasificar en la población
* Se ajusta con Laplace (sumarle 1)
* Admite variables categóricas y/o numéricas. Las categóricas las toma como factor
    * Categorias muestra probabilidades
    * Numéricas muestra valores medios o parámetros estimados

### Cómo funciona

1. **Basado en la probabilidad**: el modelo se basa en el teorema de Bayes, que ayuda a calcular la probabilidad de un evento basándose en la probabilidad de otro evento relacionado 
2. **Suposición de independencia**: la clave de Naive Bayes es la suposición de independencia condicional. Esto significa que la presencia de una característica en un dato es completamente independiente de la presencia de otras características, dado que sabemos a qué clase pertenece el dato. Por ejemplo, en un correo electrónico, la presencia de la palabra "gratis" se considera independiente de la presencia de la palabra "oferta", aunque en la realidad no sea así
3. **Cálculo de probabilidades**: para clasificar un nuevo dato, Naive Bayes calcula la probabilidad de que cada una de las clases pueda generar ese dato, basándose en las probabilidades de las características de entrenamiento 
4. **Predicción**: se selecciona la clase que tiene la mayor probabilidad para el nuevo dato

### Ventajas

* **Eficiente y rápido**: a pesar de su simplicidad, es computacionalmente eficiente y rápido de entrenar, especialmente con grandes conjuntos de datos 
* **Buen rendimiento en datos de alta dimensión**: es muy eficaz para tareas de clasificación de textos y otras aplicaciones con muchos atributos 
* **Requiere pocos datos**: solo se necesita una pequeña cantidad de datos de entrenamiento para estimar los parámetros necesarios

## KNN - K-Nearest Neighbors (K-Vecinos Más Cercanos)

### Concepto

* Es un algoritmo de **aprendizaje supervisado (se conoce la clase objetivo) no paramétrico (no construye un modelo)** que se usa para tareas de clasificación y regresión. Es simplemente aplicar un algoritmo. Es multiclase
* Su funcionamiento se basa en la idea de que los puntos de datos similares suelen tener etiquetas similares
* Para clasificar un nuevo punto, el algoritmo encuentra los "k" vecinos más cercanos en el conjunto de datos y asigna al nuevo punto la clase más común entre esos vecinos (para clasificación) o el valor promedio (para regresión

### Cómo funciona

1. **Cálculo de distancias**: se calcula la distancia entre el nuevo punto de datos y todos los puntos del conjunto de entrenamiento. La distancia euclidiana es una de las métricas más comunes, aunque existen otras como la distancia de Manhattan o la de Minkowski
2. **Identificación de vecinos**: el algoritmo selecciona los "k" puntos más cercanos al nuevo punto, basándose en la distancia calculada. El valor de "k" es un parámetro que se elige según la complejidad del problema. Se sugiere "k" impar para evitar empates
3. **Asignación de la clase**:
    * **Clasificación**: se elige la clase que más se repite entre los "k" vecinos más cercanos
    * **Regresión**: se toma el valor promedio de los "k" vecinos más cercanos para predecir el valor del nuevo punto
4. Las variables deben ser todas numéricas, aunque pueden usarse categóricas convertidas a numéricas ficticiamente (dummies)

### Características

* **No paramétrico**: no asume una distribución específica de los datos 
* **Aprendizaje instanciado o "perezoso"**: no construye un modelo explícito en el entrenamiento; almacena todo el conjunto de datos y realiza el cálculo de las distancias en el momento de hacer una predicción
* **Simple y efectivo**: competitivo con algoritmos mas complejos
* **Versátil**: Se puede usar para clasificación (asignar categorías) y regresión (predecir valores numéricos)
* **No le afecta multicolinealidad**
* **No es robusto a datos atípicos**
* **Con muchas predictoras, costoso computacionalmente**
* **"k" grandes da resultados más estables, pero mayor costo**

### Programación

| Pasos            | R                                                                                                                      |
| --               | --                                                                                                                     |
| Se crean modelos | cuantis.knn <- knn(train = select(train, !y), test = select(test, !y), cl = train\\$y, k = 3, prob = T)                |
|                  |       # predictores en train, predictores en test, clase real en entrenamiento, vecinos cercanos, calcule una prob     |
| Prob             | prob <- attr(cuantis.knn, "prob") # prob = (#casos de la clase mas votada)/k                                           |
| Matriz confusión | confusion <- table(predict = cuantis.knn, Actual = test\\$y)                                                           |
|                  | confusionMatrix(confusion, positive = "1")                                                                             |

## SVM (Máquina de Vectores de Soporte)

### Concepto

* Es un algoritmo de **aprendizaje automático supervisado (se conoce la clase)** que se utiliza para clasificación y regresión. Es multiclase
* Objetivo: encontrar el hiperplano óptimo que maximice la distancia entre diferentes clases de datos en un espacio N-dimensional, buscando un límite de decisión que funcione bien incluso con datos no linealmente separables mediante el uso de funciones kernel y vectores de soporte
    * Se busca la recta cuya mínima distancia sea máxima, es decir, que separe BIEN
    * Se elige el de mayor margen de separación: el maximal margin hyperplane
* Se ajusta con un hiperparámetro de tuning **C**
    * Con un C grande se penaliza mucho (0 error)
    * Con un C chico se penaliza poco  

### Cómo funciona

1. **Búsqueda del hiperplano**: en esencia, una SVM busca la línea (o plano en dimensiones superiores) que mejor separa (geometricamente) los puntos de datos pertenecientes a diferentes categorías 
2. **Maximización del margen**: esta línea, o hiperplano, se elige de tal forma que maximiza el margen, es decir, la distancia entre ella y los puntos más cercanos de cada clase. Estos puntos cercanos se conocen como vectores de soporte 
3. **Funciones Kernel**: para datos que no se pueden separar linealmente, las SVM utilizan funciones kernel para transformar los datos a un espacio de mayor dimensión, donde pueden volverse linealmente separables. Algunos kernels comunes incluyen el lineal, polinomial y el de base radial (RBF)

### Ventajas

* **Eficacia en alta dimensión**: son efectivos en espacios con muchas dimensiones, incluso cuando el número de dimensiones es mayor que el de las muestras 
* **Eficiencia de memoria**: utilizan un subconjunto de puntos de entrenamiento (los vectores de soporte) para la función de decisión, lo que las hace eficientes en términos de memoria 
* **Versatilidad**: la capacidad de usar diferentes funciones kernel las hace muy versátiles y adaptables a distintos tipos de datos
* **No distribución**
* **Bien con muchos atributos**
* **Robusto a datos atípicos**
* **No importa la multicolinealidad**

### Desventajas

* **No dá probabilidades**
* **Malo si hay mucho desbalance**
* **Mal si hay mucha "mezcla"**: dificil decidir el kernel
