# Puntos ideales / redes
---

## Modelo de puntos ideales (IRT - Item Response Theory)

$$Pr(y_{ij} = 1) = \Phi(\alpha_j * \theta_i - \beta_j)$$

* $\theta_i$: **punto ideal** (ubicación en el espectro ideológico)
* $\beta_j$: **punto de corte** (posición ideológica donde Pr(Sí) = 0.5
* $\alpha_j$: **discriminación** (qué tan bien separa la votación)
* **IRT**: modelo probabilístico que estima la posición ideológica latente de cada legislador. Reduce dimensionalidad

## Análisis de puntos ideales

* **Objetivo**: Ubicar a cada diputado en un espectro ideológico unidimensional basándose en sus patrones de votación en el Congreso

* **Pipeline**:
1. **Web Scraping**: extracción de datos de votaciones desde HCDN
2. **Limpieza y Transformación**: normalización de nombres, estandarización de formatos
3. **Contrucción de Matriz**: Legisladores x Actas. Codificación 1- NEG, 2- AFI, 3-ABS, 4-AUS
4. **Modelado Estadístico**: algoritmo IRT para cálculo de puntos ideales
5. **Visualización y Análisis**: interpretación de resultados y distancias políticas 

## Redes

* Grafo de una red G<V, A>
* Propiedades:
    * dirigidas vs no dirigidas
    * grado de un nodo: número de conexiones
* Matriz de adyacencia vs Lista de adyacencia
* Distancia de un grafo: camino más corto
* Betweeness: número de visitas que tendrá un nodo, al recorrer todos los caminos a través de los caminos más cortos
* Influencia colectiva: medida del número de nodos a los que un nodo en particular puede influenciar, directa o indirectamente
* Transitivity: mide la probabilidad de que dos nodos conectados a un mismo nodo también estén conectados entre sí
* Comunidad: grupos de nodos densamente conectados entre sí
