## Estructura HTML

* **DOM**: Document Object Model
* Es la representación estructurada de un documento HTML, como un árbol de objetos
* **Elementos claves**:
    * Tags: \<div>, \<p>, \<h1>, etc.
    * Atributos: class, id, href, etc.
    * Jerarquía: elementos padre, hijo, hermanos
    * Contenido: texto dentro de los elementos

```html
<!DOCTYPE html>
<html>
    <head>
        <title>Mi Página</title>
    </head>
    <body>
        <div class="contenedor">
            <h1 id="titulo">Título Principal</h1>
            <p class="descripcion">Este es un párrafo</p>
            <ul>
                <li class="item">Elemento 1</li>
                <li class="item">Elemento 2</li>           
            </ul>
        </div>
    </body>
</html>
```

## Selectores XPath vs CSS

* **Selectores CSS**: simple y efectivo

```
/* Por tag */
p

/* Por clase */
.mi-clase

/* Por ID */
#mi-id

/* Combinaciones */
div.contenedor p.descripcion

/* Atributos */
a[href*="github"]

/* Pseudo-selectores */
li:first-child
tr:nth-child(2n)
```

* **Selectores xpath**: más potente, más complejo

```
// Ruta absoluta
/html/body/div/p

// Ruta relativa
//div[@class="contenedor"]//p

// Por texto
//a[text()="Siguiente"]

// Contiene texto
//span[contains(text(), "precio")]

// Por posición
//li[1] // Primer elemento
//li[last()] // Último elemento

// Operadores lógicos
//div[@class="item" and @id="especial"]
//p[contains(@class, "destacado") or @class="importante"]

// Navegación de ejes
//h1/following-sibling::p // Hermanos siguientes
//td/parent::tr // Elemento padre
//div/descendant::a // Todos descendientes
```

| CSS Selector                      | XPath                         |
| --                                | --                            |
| Sintaxis simple y clara           | Muy potente y flexible        |
| Más rápido en navegadores         | Navegación bidireccional      |
| Familiar para desarrolladores web | Búsquedas complejas por texto |
| Limitado para navegación compleja | Curva de aprendizaje más alta |

* Recomendación: CSS para casos simples, XPath para lógica compleja

* **Python**: request + BeatifulSoup, scrapy, httpx

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd

response = requests.get("https://quotes.toscrape.com/")
soup = BeautifulSoup(response.content, 'html.parser')
quotes = []
for quote in soup.find_all('div', class_='quote'):
    text = quote.find('span', class_='text').text
    author = quote.find('small', class_='author').text
    quotes.append({'texto': text, 'autor': author})

df = pd.DataFrame(quotes)
df.head()
````

## Ética y Legalidad

* **Aspectos Legales**
    * robots.txt: Siempre revisar /robots.txt
    * Términos de Servicio: Leer ToS del sitio
    * Rate Limiting: No sobrecargar servidores
    * Copyright: Respetar propiedad intelectual
* **Aspectos Éticos**
    * Propósito legítimo: Investigación, periodismo
    * Datos públicos: Solo info accesible públicamente
    * Privacidad: No datos personales sin consentimiento
    * Transparencia: Identificarse como bot

El scraping responsable protege tanto al usuario como al sitio web

````python
import time
import random
import requests

def scrape_responsibly(urls):
    for url in urls:
        # Delay aleatorio entre requests (1-3 segundos)
        time.sleep(random.uniform(1, 3))
        # Headers que simulan navegador real
        headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
        try:
            response = requests.get(url, headers=headers, timeout=10)
            # Procesar datos...
        except requests.exceptions.RequestException as e:
            print(f"Error en {url}: {e}")
            continue
````

## Casos Prácticos Regionales

* Datasets Argentinos:
    * [Datos Argentina](https://datos.gob.ar/)
    * [INDEC](https://indec.gob.ar/)
    * [BCRA](https://bcra.gob.ar/)

````python
import requests
import pandas as pd

def obtener_datos_indec():
    # API del INDEC para índices económicos
    url = "https://apis.datos.gob.ar/series/api/series"
    params = {
        'ids': 'tcrse_2weZeH', # IPC
        'format': 'json'
    }
    response = requests.get(url, params=params)
    data = response.json()
    df = pd.DataFrame(data['data'])
    return df

# Análisis de inflación
inflacion = obtener_datos_indec()
print(inflacion.head())
````
