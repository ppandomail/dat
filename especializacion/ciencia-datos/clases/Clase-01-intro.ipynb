## Introducción

* **Las máquinas, nos van a reemplazar?**
    * En tareas automáticas, repetitivas
* **Póker**: Tensión entre Estrategias Metódicas/Algorítmicas vs Juego de Explotación Humana (ej. microgestos)
* **Vibe coding**: un método de desarrollo de software popularizado a principios de 2025 por Andrej Karpathy que utiliza modelos de lenguaje grandes (LLM) para generar código a partir de indicaciones en lenguaje natural
* **Sesgo**: Error que produce una distorsión en los resultados de un estudio. Se trata de una influencia sistemática que da lugar a una sobrestimación o infraestimación del verdadero valor de la población

## Contexto de aplicación

* **Explicar** la realidad: análisis exploratorio, segmentación, filtros, zoom in/ zoom out, visualización
* **Predecir** mediante modelos (correlaciones, tendencias). Generar datos y modelado predictivo. Clasificación / Regresión
* **Decidir**: toma de decisiones
    * Inferencia causal: permite usar datos no generados y estimar el impacto menos sesgada
    * Toma de decisiones asistida por escenarios alternativos (que hubiera pasado si...)

1. Entrar a [Claude AI](https://claude.ai/download)
2. Descargarlo según el Sistema Operativo
3. Una vez instalado, buscar el archivo claude_desktop_config.json y reemplazar por esto (asegurarse de escribir los path a donde se quiere que Claude acceda)

```
{
  "mcpServers": {
    "filesystem": {
      "command": "npx",
      "args": [
        "-y",
        "@modelcontextprotocol/server-filesystem",
        "/Users/username/Desktop",
        "/path/to/other/allowed/dir"
      ]
    }
  }
}
```
4. Salir de claude y volver a entrar (para que tome los cambios)
5. Probar por ejemplo escribiendole: crear un archivo de texto con el nombre prueba.txt
6. Otra prueba: escribirle que explique la ley de innovación tecnológica en graphviz en un archivo .dot
