## Preguntas de interés

1. **Predicción de voto**
    * ¿Cómo votará un diputado en un proyecto futuro según su historial?
    * ¿Qué características de un proyecto (tema, tipo, autoría) predicen mayor probabilidad de aprobación?

2. **Clasificación de afinidades políticas**
    * ¿Qué diputados tienen patrones de votación similares?
    * ¿Se puede agrupar a los diputados por bloques ideológicos basados en sus votos?

3. **Análisis de temas y tendencias**
    * ¿Qué temas tienen mayor tasa de aprobación o rechazo?
    * ¿Cómo evolucionan las votaciones sobre temas como salud, educación o economía a lo largo del tiempo?

4. **Detección de anomalías**
    * ¿Qué votaciones se apartan del comportamiento habitual de un diputado o bloque?
    * ¿Cuáles son los proyectos que generan divisiones internas en bloques?

5. **Modelos de recomendación**
    * ¿Qué proyectos podrían interesar a un diputado según sus votos anteriores?
    * ¿Qué tipo de dictámenes suelen recibir apoyo transversal?

6. **Redes**
    * Analizar la votación de los diputados de cada provincia para detectar la afinidad entre ellas (teniendo o no en cuenta los bloques)
    * Analizar la afinidad de bloques en cada provincia
    * Descubrir el bloque mas influyente en cada provincia
