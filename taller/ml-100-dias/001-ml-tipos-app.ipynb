# 001 - ML, Tipos y Aplicaciones
---

## ML

* Es una rama de la IA
* Permite a los sistemas aprender de los datos, tomar decisiones y mejorar con el tiempo sin necesidad de programación explícita
* Los algoritmos análizan patrones en los datos para identificar la mejor manera de alcanzar una solución

## Tipos



|                    |                                                                                                  |
| --                 | --                                                                                               |
| **Supervisado**    | el algoritmo aprende de un conjunto de datos etiquetado                                          |
| **No Supervisado** | el algoritmo trabaja con datos sin etiquetar. Encontrar patrones, correlaciones o grupos ocultos |
| **Por refuerzo**   | el algoritmo aprende por ensayo y error. Retroalimentación pos+ (recompensas), neg- (castigos)   |

## Aplicaciones

|                 |                                                                      |
| --              | --                                                                   |
| **Salud**       | diagnosticar enfermedades, predecir y monitorear tendencias de salud |
| **Finanzas**    | detectar fraudes, gestionar riesgos, decisiones comerciales          |
| **Comercio**    | recomendaciones, stock, optimizar logística                          |
