# 007 - Fundamentos de ingeniería de características: selección y extracción de características
---

* Las características (columnas o atributos) representan la materia prima para realizar predicciones
* Refinar, seleccionar e incluso crear nuevas características para mejorar el rendimiento del modelo

## Selección

* Al reducir los datos irrelevantes o redundantes, se minimiza el sobreajuste, se simplifica el modelo y se mejora la interpretabilidad
* Métodos:
    * **Método de filtro**: clasifican las características según medidas estadísticas. Por ejemplo, se pueden calcular coeficientes de correlación o usar pruebas estadísticas como ANOVA para identificar las características más relevantes para la variable objetivo
    * **Métodos de envoltura**: evalúan todas las combinaciones posibles de características para encontrar el mejor subconjunto. Algunos ejemplos son la selección hacia adelante y la eliminación hacia atrás .
    * **Métodos integrados**: se integran en algoritmos que realizan la selección de características durante el proceso de entrenamiento del modelo. La regresión Lasso, por ejemplo, reduce las características menos relevantes a cero, seleccionando eficazmente las más importantes

## Extracción

* Crea nuevas características o transforma las existentes
* Métodos:
    * **Análisis de Componentes Principales (PCA)**: reduce la dimensionalidad del conjunto de datos, conservando al máximo la varianza. Al transformar las características correlacionadas en un conjunto de componentes principales no correlacionados, el PCA captura la estructura subyacente de los datos en menos dimensiones
    * **Transformaciones específicas del dominio**: En ocasiones, la experiencia en el dominio facilita la creación de nuevas características a partir de las existentes. Por ejemplo, si se predice el precio de las viviendas, combinar la superficie y el número de dormitorios en una nueva característica como "habitaciones por metro cuadrado" podría aportar información valiosa
    * **Características polinómicas**: crean nuevas características al elevar las existentes a una potencia. Si existe una relación no lineal entre una característica y el objetivo, las características polinómicas pueden ayudar a modelarla con mayor eficacia

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
import numpy as np

# Sample data
data = {
    'square_footage': [1500, 1800, 2400, 3000, 3500],
    'num_bedrooms': [3, 4, 3, 5, 4],
    'num_bathrooms': [2, 2, 3, 3, 4],
    'location_rating': [8, 7, 9, 6, 7]
}
df = pd.DataFrame(data)

# Feature Selection: Select 'location_rating' and 'square_footage' based on correlation analysis
selected_features = df[['location_rating', 'square_footage']]

# Feature Extraction: Create new feature 'rooms_per_sqft'
df['rooms_per_sqft'] = df['square_footage'] / (df['num_bedrooms'] + df['num_bathrooms'])

# Add Polynomial Feature for 'square_footage'
poly = PolynomialFeatures(degree=2, include_bias=False)
square_footage_poly = poly.fit_transform(df[['square_footage']])
df['square_footage_squared'] = square_footage_poly[:, 1]

df[['square_footage', 'num_bedrooms', 'num_bathrooms', 'location_rating', 'rooms_per_sqft', 'square_footage_squared']]
```
