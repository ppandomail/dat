* Ayudan a evaluar el rendimiento del modelo, especialmente en tareas de clasificación binaria

## Precisión y recuperación

* Matriz de confusión
* **Verdadero positivo (VP)**
* **Verdadero negativo (VN)**
* **Falso positivo (FP)**
* **Falso negativo (FN)**

| Actual / Predicho | Positivo | Negativo |
| --                | --       | --       |
| Positivo          | VP       | FP       |
| Negativo          | FN       | VN       |

* **Precisión**: cuando el modelo predice algo positivo, ¿con qué frecuencia es correcto?
    * Alta precisión significa pocos falsos positivos 

$$Precisión = \dfrac{VP}{VP + FP}$$

* **Recordar**: Cuando el resultado real es positivo, ¿con qué frecuencia el modelo lo predice correctamente?
    * Alto nivel de recuperación significa pocos falsos negativo

$$Recall = \dfrac{VP}{VP + FN}$$

## F1

* Combina la precisión y la recuperación en una sola métrica calculando su media armónica
    * F1 alta indica un buen equilibrio entre precisión y recuperación 

$$F1 = 2 * \dfrac{Precisión * Recall}{Precisión + Recall}$$

## Curva ROC (Receiver Operating Characteristic)

* Visualiza el rendimiento del modelo en diferentes umbrales, trazando la tasa de positivos verdaderos (TVP) frente a la tasa de falsos positivos (TFP)

$$TVP = \dfrac{VP}{VP + FN}$$
$$TFP = \dfrac{FP}{FP + VN}$$

## AUC (Area Under Curve)

* Valor para evaluar la capacidad del modelo para distinguir entre clases
* Cuanto más cercano esté el AUC a 1, mejor será el rendimiento del modelo entodos los umbrales

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Synthetic true labels and model scores (probabilities)
y_true = [0, 0, 1, 1]
y_scores = [0.1, 0.4, 0.35, 0.8]

# Calculate ROC curve
fpr, tpr, thresholds = roc_curve(y_true, y_scores)

# Calculate AUC
auc_score = roc_auc_score(y_true, y_scores)

# Plot ROC Curve
plt.plot(fpr, tpr, label=f'AUC = {auc_score:.2f}')
plt.plot([0, 1], [0, 1], 'k--')  # Random guessing line
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()
```

En este gráfico, la curva muestra la precisión con la que el modelo separa las dos clases en distintos umbrales. Una curva ROC más pronunciada y un AUC alto indican un mejor rendimiento
