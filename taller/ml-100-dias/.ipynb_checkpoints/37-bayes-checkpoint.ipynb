* Es un clasificador probabilistico basado en el teorema de Bayes
* Eficaz para ciertas tareas como la detección de spam, análisis de sentimientos, sistemas de recomendación

## Teorema de Bayes

* Calcular la probabilidad de una hipótesis dada cierta evidencia (probabilidad condicional de un evento aleatorio)

$$P(A|B) = \frac{P(B|A) * P(A)}{P(B)}$$

* $P(A|B)$: Probabilidad de que el evento A ocurra dado que el evento B ha ocurrido

## Supuestos de Naive Bayes

1. Independencia de las características
1. Independencia condicional: cada característica contribuye independientemente a la probabilidad de la clase objetivo, dada la etiqueta de clase

## Pasos en Naive Bayes

* 1. Calcular las probabilidades previas $P(A)$ para cada clase (por ejemplo, spam o no spam)
* 2. Calcular la probabilidad $P(B∣A)$ para cada característica
* 3. Utilizar el teorema de Bayes para calcular la probabilidad posterior de cada clase
* 4. Asignar la clase con la mayor probabilidad posterior

## Ejemplo matemático: predecir si un email es spam (S) o no (N) 

| Feature | free | offer | win | spam? |
| --      | --   | --    | --  | --    |
| email1  | Yes  | Yes   | No  | Yes   |
| email2  | No   | Yes   | Yes | Yes   |
| email3  | No   | No    | Yes | No    |
| email4  | Yes  | No    | No  | No    |

* $P(S)$ = 2/4 = 0.5
* $P(N)$ = 2/4 = 0.5
* $P(free|S)$ = 1/2 = 0.5
* $P(free|N)$ = 1/2 = 0.5


## Casos de uso ideales para Naive Bayes

1. Clasificación de texto
2. Diagnóstico médico
3. Categorización de documentos
4. Detección de fraude

## Ejemplo Naive Bayes para análisis de sentimientos

* Clasificación de las reseñas de películas como positivas o negativas

### Paso 1: Importar bibliotecas y cargar datos 

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Load dataset
df = pd.read_csv('data/IMDB_Dataset.csv')
df.head()
```

* **review**: reseña textual de la película
* **sentiment**: sentimiento asociado con la reseña (positive o negative)

### Paso 2: Preprocesamiento de texto

```python
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

# Clean text data
stemmer = PorterStemmer()
#stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    text = re.sub('[^a-zA-Z]', ' ', text)  # Remove special characters
    text = text.lower()                    # Convert to lowercase
    text = text.split()                    # Tokenize
    #text = [stemmer.stem(word) for word in text if word not in stop_words]
    return ' '.join(text)

df['review'] = df['review'].apply(preprocess_text)
```

* Los datos de texto sin procesar contienen ruido, como caracteres especiales, puntuación y palabras irrelevantes
* El preprocesamiento limpia los datos, haciéndolos aptos para el modelado

### Paso 3: Convertir texto en características

```python
# Convert text to numeric features
# conserva solo las 5000 palabras más frecuentes, lo que reduce la carga computacional
vectorizer = CountVectorizer(max_features=5000)
X = vectorizer.fit_transform(df['review']).toarray()
y = df['sentiment'].map({'positive': 1, 'negative': 0})
```

### Paso 4: División de entrenamiento y prueba

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

### Paso 5: Entrenar el modelo Naive Bayes

```python
# Train Multinomial Naive Bayes (ideal para clasificación de texto)
model = MultinomialNB()
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)
```

### Paso 6: Evaluar el modelo

```python
# Evaluation: métricas para evaluar el rendimiento del modelo
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```
