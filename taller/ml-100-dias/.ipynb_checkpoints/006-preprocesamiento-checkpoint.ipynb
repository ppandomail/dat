# 006 Técnicas de preprocesamiento de datos
---

* Sin datos bien preprocesados, incluso los mejores algoritmos pueden producir resultados deficientes
* Los datos sin procesar suelen contener inconsistencias, ruido y escalas variables que dificultan la generalización de los modelos

## Normalización

* Escala las características para que se encuentren dentro de un rango específico, generalmente de 0 a 1
* Útil cuando se conocen los límites mínimos y máximos de los datos o cuando se trata con modelos sensibles a la magnitud de los datos

$$X_{norm} = \dfrac{X - X_{min}}{X_{max} - X_{min}}$$

## Estandarización

*  Transforma los datos para que tengan una media de 0 y una desviación estándar de 1
*  A diferencia de la normalización, que está limitada a un rango específico, la estandarización reconfigura la distribución de los datos alrededor de cero, lo que la hace adecuada para datos sin límites predefinidos
*  Útil para los datos que tienen una distribución gaussiana (normal) y se aplica comúnmente en algoritmos que suponen normalidad, como la regresión lineal, la regresión logística y las redes neuronales

$$X_{std} = \dfrac{X - mean(X)}{std(X)}$$

## Codificación

* Para datos categóricos, las transformaciones numéricas son necesarias, ya que los modelos de aprendizaje automático no procesan directamente las etiquetas de texto
* **Codificación One-Hot**: Crea una columna binaria para cada categoría, donde un "1" representa la presencia de dicha categoría. Resulta útil para datos categóricos desordenados
* **Codificación de etiquetas**: Asigna a cada categoría un entero único. Sin embargo, solo es adecuado cuando existe un orden significativo en las categorías (como bajo, medio, alto)

Ejemplo : Supongamos que se tiene una columna "color" con valores ["Rojo", "Azul", "Verde"]. Con la codificación one-hot, esto se convertiría en tres columnas binarias:

| Red | Blue | Green |
| --  | --   | --    |
| 1   | 0    | 0     |
| 0   | 1    | 0     |
| 0   | 0    | 1     |

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, OneHotEncoder

# Sample data
data = {
    'item_price': [100, 150, 200, 250, 300],
    'quantity_sold': [1, 3, 2, 5, 4],
    'store_location': ['New Delhi', 'Mumbai', 'Chennai', 'Kolkata', 'New Delhi']
}
df = pd.DataFrame(data)

# Normalization using MinMaxScaler
scaler = MinMaxScaler()
df[['item_price', 'quantity_sold']] = scaler.fit_transform(df[['item_price', 'quantity_sold']])

# Standardization using StandardScaler
std_scaler = StandardScaler()
df['item_price_std'] = std_scaler.fit_transform(df[['item_price']])

# Encoding categorical data using OneHotEncoder
encoder = OneHotEncoder(sparse_output=False)
location_encoded = encoder.fit_transform(df[['store_location']])
location_df = pd.DataFrame(location_encoded, columns=encoder.get_feature_names_out(['store_location']))

# Concatenate encoded columns with original data
df = pd.concat([df, location_df], axis=1)

# Drop original categorical column
df = df.drop('store_location', axis=1)

print(df)
```

* Estas 3 técnicas transforman los datos a un formato listo para el modelado.
* Al preprocesar los datos eficazmente, se garantiza que los modelos se centren en patrones en lugar de inconsistencias de datos o problemas de escala, ayudando a que mejoren la precisión y aumenten la interpretabilidad
