## Probabilidad básica

* Cuantifica la probabilidad de que ocurra un evento
* Para un evento dado, la probabilidad se puede calcular como

$$P = \dfrac{casos \, favorables}{total \, de \, casos}$$

* Por ej. si se lanza un dado de seis caras, la probabilidad de sacar un 3 es 1/6

## Probabilidad condicional

* Es la probabilidad de que un evento ocurra dado que otro ya ha ocurrido
* Es útil cuando se quiere calcular la probabilidad de un evento específico con información adicional

$$P(A|B) = \dfrac{P(A \, and \, B)}{P(B)}$$

* Por ej.
* Número total de cartas en una baraja: 52
* Número de cartas rojas en una baraja (corazones y diamantes): P(B): 26
* Número de corazones en una baraja: P(A y B): 13
* La probabilidad de sacar un corazón dado que la carta es roja se calcula

$$P(Corazón|Rojo) = \dfrac{Corazones}{Rojas} = \dfrac{13}{26} = 0.5$$

## Teorema de Bayes

* Permite invertir las probabilidades condicionales

$$P(A|B) = \dfrac{P(B|A) P(A)}{P(B)}$$

* En ML, el teorema de Bayes se utiliza en algoritmos, para tareas de clasificación como la detección de spam o el diagnóstico médico, donde se necesita revisar las estimaciones de probabilidad en función de los datos observados

```python
# Given probabilities
p_spam = 0.20                             # Probability of any email being spam
p_not_spam = 1 - p_spam                   # Probability of an email being not spam
p_spam_notification_given_spam = 0.95     # Probability that spam is correctly marked as spam
p_spam_notification_given_not_spam = 0.05 # False positive rate

# Total probability of a spam notification
p_spam_notification = (p_spam_notification_given_spam * p_spam) + (p_spam_notification_given_not_spam * p_not_spam)

# Bayes' Theorem: Probability of an email actually being spam given it was marked as spam
p_spam_given_spam_notification = (p_spam_notification_given_spam * p_spam) / p_spam_notification

print("Probability that an email is actually spam given a spam notification:", p_spam_given_spam_notification)
```

**Si un email se marca como spam, existe un 83% de probabilidad de que sea spam, según la precisión de nuestro filtro y la proporción general de spam. Esta métrica es útil para diseñar y evaluar sistemas de detección de spam, ya que considera no solo la precisión del filtro, sino también la frecuencia real de los spam**
