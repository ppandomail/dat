# 010 - Métricas de evaluación para la regresión
---

* ¿Cómo saber si el modelo es preciso?
* Las métricas de evaluación ayudan a cuantificar la precisión de estas predicciones e identificar áreas de mejora
* La elección de la métrica de evaluación adecuada depende de lo que se intente medir:
    1. ¿Qué tan cerca están las predicciones de los valores reales?
    2. Si el modelo predice consistentemente en exceso o en defecto
    3. Qué tan sensible es la métrica a los valores atípicos

## Error cuadrático medio (MSE)

* Es el promedio de las diferencias al cuadrado entre los valores reales y los previstos
* Elevar al cuadrado los errores penaliza más los errores grandes que los pequeños, lo que hace que el MSE sea sensible a los valores atípicos

$$MSE = \dfrac{1}{n} \Sigma^n_{i=1} (y_i - \hat y_i)^2$$

* Un MSE más bajo indica un mejor rendimiento del modelo
* **Ventajas**: penaliza errores grandes, lo que lo hace útil para detectar desviaciones significativas
* **Contras**: debido a que los errores se elevan al cuadrado, MSE puede ser más sensible a los valores atípicos de lo deseado

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error (MSE):", mse)
```

## Error absoluto medio (MAE)

* Mide la magnitud promedio de los errores en un conjunto de predicciones, sin considerar su dirección
* Es menos sensible a los valores atípicos

$$MAE = \dfrac{1}{n} \Sigma^n_{i=1}|y_i - \hat y_i|$$

* **Ventajas**: proporciona una interpretación sencilla de la magnitud del error
* **Contras**: trata todos los errores por igual, por lo que los errores grandes no se penalizan más que los más pequeños

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)
print("Mean Absolute Error (MAE):", mae)
```

## Error cuadrático medio (RMSE)

* Es la raíz cuadrada del MSE, da el error en las mismas unidades que la variable objetivo
* Combina las ventajas del MSE (sensibilidad a grandes errores) con la interpretabilidad, ya que se mide en las mismas unidades que los datos originales

$$RMSE = \sqrt{\dfrac{1}{n} \Sigma^n_{i=1} (y_i - \hat y_i)^2}$$

* Un RMSE más bajo indica un mejor ajuste y, como está en las mismas unidades que el objetivo, es más intuitivo que MSE
* **Ventajas**: combina la interpretabilidad de MAE con la sensibilidad a valores atípicos de MSE
* **Contras**: puede verse muy influenciado por valores atípicos, al igual que MSE

```python
import numpy as np

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("Root Mean Squared Error (RMSE):", rmse)
```

## Puntuación $R^2$ (coeficiente de determinación)

* Indica la proporción de varianza en la variable dependiente que es predecible a partir de la(s) variable(s) independiente(s)
* En esencia, indica qué tan bien el modelo explica la variabilidad en los datos objetivo

$$R^2 = 1 - \dfrac{\Sigma (y_i - \hat y_i)^2}{\Sigma (y_i - \bar y_i)^2}$$

* R² = 1 : Ajuste perfecto
* R² = 0 : El modelo no funciona mejor que una media simple
* R² < 0 : El modelo funciona peor que la media
* **Ventajas**: proporciona una medida normalizada (entre 0 y 1) del rendimiento del modelo
* **Contras**: agregar características irrelevantes a veces puede inflar el R², dando una sensación engañosa de mejora

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
print("R² Score:", r2)
```
