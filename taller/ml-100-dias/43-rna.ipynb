# 043 - Redes Neuronales Artificiales (RNA)
---

* Intentan imitar el funcionamiento del cerebro para resolver problemas complejos
* Son la base de muchas aplicaciones avanzadas de aprendizaje automático, desde el reconocimiento de imágenes hasta el procesamiento del lenguaje natural

## Perceptrón

* Es un modelo matemático simplificado de una neurona biológica
* **Entradas**: una neurona artificial recibe entradas numéricas (por ejemplo, valores de píxeles de una imagen)
* **Pesos**: cada entrada se multiplica por un peso, que representa la fuerza de la conexión
* **Suma**: se suman las entradas ponderadas
* **Función de activación**: sa suma pasa a través de una función de activación (como sigmoide o ReLU) para determinar si la neurona debe "dispararse" (emitir un valor)
* **Salida**: la salida final se pasa a la siguiente capa de neuronas

## Aplicaciones de las RNA

* **Reconocimiento de imágenes**: se utiliza en imágenes médicas, reconocimiento facial y vehículos autónomos
* **Procesamiento de voz**: potenciando asistentes virtuales como Siri y Alexa
* **Procesamiento del lenguaje natural**: habilitación de traducción, chatbots y análisis de sentimientos
* **Sistemas de recomendación**: impulsar recomendaciones personalizadas en plataformas como Netflix y Amazon
* **Finanzas**: detección de fraude, predicción de precios de acciones y comercio algorítmico
