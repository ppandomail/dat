* Estas métricas facilitan la selección de los algoritmos adecuados, optimizan el rendimiento del modelo y comunican los hallazgos con claridad

## Tendencia central

* Se refiere a un valor único que intenta describir el conjunto de datos
* Intenta referirse al centro de la distribución de datos
* La media, la mediana y la moda normalmente se denominan medidas de tendencia central y se utilizan a menudo en estadística



## Media (o promedio)

* Es la suma de todos los valores de un conjunto de datos dividida entre el número de valores

## Mediana

* Es el valor central de un conjunto de datos cuando los valores se ordenan de forma ascendente o descendente
* Si el número de valores es impar, la mediana es el valor central
* Si el número de valores es par, es el promedio de los dos valores centrales
* Para los datos [10, 20, 30, 40, 50], la mediana es 30
* Para los datos [10, 20, 30, 40], la mediana es (20 + 30)/2 = 25

## Moda

* Es el valor que aparece con mayor frecuencia en un conjunto de datos
* Algunos conjuntos de datos pueden no tener moda si todos los valores son únicos, mientras que otros pueden ser multimodales si tienen más de una moda
* Para los datos [10, 15, 15, 20, 25], la moda es 15
* Para los datos [5, 10, 10, 15, 15, 20], hay dos modas: 10 y 15

## Varianza

* Mide la distancia entre cada valor del conjunto de datos y la media
* Una varianza más alta indica que los puntos de datos están más dispersos
* Una varianza más baja significa que están más cerca de la media
* En ML, la varianza ayuda a comprender la variabilidad de los datos y la forma de su distribución

## Desvio estándar

* Es la raíz cuadrada de la varianza
* Proporciona una medida de dispersión en la misma unidad que los datos originales
* Al igual que la varianza, ayuda a indicar cuán concentrados o dispersos están los valores con respecto a la media

```python
import numpy as np
from scipy import stats

# Test scores of students
scores = [75, 85, 90, 95, 85, 80, 78, 85, 92, 88]

# Mean
mean_score = np.mean(scores)
print("Mean:", mean_score)

# Median
median_score = np.median(scores)
print("Median:", median_score)

# Mode
mode_score = stats.mode(scores)
print("Mode:", mode_score[0])

# Variance
variance_score = np.var(scores)
print("Variance:", variance_score)

# Standard Deviation
std_deviation_score = np.std(scores)
print("Standard Deviation:", std_deviation_score)
```

**Estos resultados indican que la puntuación media ronda los 85, y la mayoría de los estudiantes obtienen puntuaciones cercanas a este valor (como lo refleja la baja desviación estándar). La moda muestra que 85 fue la puntuación más común entre los estudiantes**
